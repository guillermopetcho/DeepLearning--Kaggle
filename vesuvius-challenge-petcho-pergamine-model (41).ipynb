### 📊 **Resumen General de la Competencia**

La **Vesuvius Challenge – Surface Detection** es una competencia de visión por computadora y aprendizaje profundo (deep learning) organizada en Kaggle con un enfoque aplicado histórico-científico: **virtualmente “desenrollar” y reconstruir los rollos de papiro carbonizados por la erupción del Monte Vesuvius hace casi 2000 años**. ([vesuviuschallenge.org][1])

En términos concretos, el objetivo es:

**👉 Entrenar un modelo que pueda detectar con alta precisión las superficies de las hojas de papiro dentro de volúmenes 3D de tomografías computarizadas (CT scans).**
Estos papeles están fuertemente distorsionados, plegados y dañados, por lo que la detección de superficies es difícil incluso para métodos clásicos. ([scrollprize.substack.com][2])

---

### 📦 **¿Qué es lo que se predice?**

Se trabaja con **volúmenes 3D de CT scans** (es decir imágenes volumétricas donde cada voxel proviene de una rebanada del escáner). Dentro de estos volúmenes se quiere **predecir qué regiones del voxel representan la *superficie del papiro*** — es decir, dónde se encuentra la lámina de la hoja en 3D. ([scrollprize.substack.com][2])

* La salida del modelo **no es simplemente una máscara 2D**, sino un **modelo 3D de la superficie** (o múltiples láminas) que recorre las estructuras internas. ([scrollprize.substack.com][2])
* Esta superficie es muy fina en Z, por lo que el modelo debe ser capaz de **detectar sub-slices o subniveles con precisión** para describir correctamente la geometría. ([scrollprize.substack.com][2])

---

### 📈 **Métrica de evaluación (según discusiones y publicaciones)**

El ranking de la competencia no se basa en una métrica simple como Dice o F1 estándar (como en muchos retos de segmentación). En Surface Detection hay **componentes adicionales relacionados con topología y conectividad de la superficie**:

✔ Existe una combinación de:

* **Superficie Dice (o similar)**: qué tan bien los voxels predichos coinciden con la superficie real.
* **Topología / conectividad**: penaliza predicciones que:

  * *rompen* la continuidad de una hoja
  * *fusionan* hojas separadas
  * producen *huecos o artefactos* no físicamente reales

👉 Esto significa que el modelo no solo debe acertar los voxels correctos, sino también producir superficies **coherentes y físicamente plausibles**. ([scrollprize.substack.com][2])

```python
# =================================================================================
# CELL 3 — LECTURA DE VOLÚMENES 3D (I/O ROBUSTO & LAZY LOADING)
# VESUVIUS CHALLENGE — SURFACE DETECTION
# =================================================================================

from pathlib import Path
from typing import Optional, Tuple, Union, Any
import numpy as np
import tifffile
import zarr
import gc
import os

# -----------------------------------------------------------------------------
# CONFIGURACIÓN GLOBAL DE I/O
# -----------------------------------------------------------------------------
# Nota: Aunque definimos float32, la carga inicial se mantiene en uint16 (raw)
# para ahorrar RAM. El casting se hace en el Dataset (on-the-fly).
IO_DTYPE = np.float32 
EXPECTED_DIM = 3   # (Z, H, W)

# -----------------------------------------------------------------------------
# VALIDADORES (LIGEROS)
# -----------------------------------------------------------------------------
def _validate_volume_shape(shape: tuple, volume_id: str):
    """
    Valida dimensiones sin leer el contenido del array.
    """
    if len(shape) != EXPECTED_DIM:
        raise ValueError(
            f"Volumen {volume_id} inválido. Esperado {EXPECTED_DIM}D, recibido {len(shape)}D. Shape: {shape}"
        )

# -----------------------------------------------------------------------------
# LECTURA TIF 3D (MEMMAP - LAZY)
# -----------------------------------------------------------------------------
def load_volume_tif(
    tif_path: Union[str, Path],
    volume_id: str
) -> Any:
    """
    Usa Memory Mapping para acceder al TIF sin cargarlo todo en RAM.
    """
    tif_path = Path(tif_path)

    if not tif_path.exists():
        raise FileNotFoundError(f"TIF no encontrado: {tif_path}")

    # Memmap permite al OS gestionar la memoria. Solo lee lo que se pide.
    volume = tifffile.memmap(tif_path)
    
    _validate_volume_shape(volume.shape, volume_id)

    return volume

# -----------------------------------------------------------------------------
# LECTURA ZARR (LAZY NATIVE)
# -----------------------------------------------------------------------------
def load_volume_zarr(
    zarr_path: Union[str, Path],
    volume_id: str
) -> Any:
    """
    Retorna el objeto Zarr directamente (es lazy por defecto).
    """
    zarr_path = Path(zarr_path)

    if not zarr_path.exists():
        raise FileNotFoundError(f"ZARR no encontrado: {zarr_path}")

    # mode='r' asegura acceso de solo lectura
    volume = zarr.open(str(zarr_path), mode="r")

    _validate_volume_shape(volume.shape, volume_id)

    return volume

# -----------------------------------------------------------------------------
# API UNIFICADA DE CARGA (LAZY)
# -----------------------------------------------------------------------------
def load_volume_3d(
    volume_info: dict,
    volume_id: str,
    prefer: str = "zarr"
) -> Any:
    """
    Carga un volumen 3D en modo Lazy (puntero a disco).
    NO realiza casting a float32 aquí para ahorrar memoria.
    """

    # 1. Intentar cargar ZARR si se prefiere y existe
    if prefer == "zarr" and volume_info.get("zarr") is not None:
        path = volume_info["zarr"]
        # Pequeña validación extra por si la ruta es string
        if isinstance(path, str): path = Path(path)
        
        if path.exists():
            return load_volume_zarr(path, volume_id)
    
    # 2. Fallback a TIF
    if volume_info.get("tif") is not None:
        path = volume_info["tif"]
        if isinstance(path, str): path = Path(path)
        
        if path.exists():
            return load_volume_tif(path, volume_id)

    raise FileNotFoundError(
        f"No hay fuente válida (TIF/ZARR) accesible para volumen {volume_id}"
    )

# -----------------------------------------------------------------------------
# CARGA SEGURA
# -----------------------------------------------------------------------------
def safe_load_volume(
    volume_info: dict,
    volume_id: str,
    prefer: str = "zarr"
) -> Any:
    try:
        volume = load_volume_3d(
            volume_info=volume_info,
            volume_id=volume_id,
            prefer=prefer
        )
        return volume
    except Exception as e:
        raise RuntimeError(f"Error cargando volumen {volume_id}: {e}")

# -----------------------------------------------------------------------------
# HELPERS DE DEBUG
# -----------------------------------------------------------------------------
def inspect_volume(volume: Any) -> dict:
    # Nota: Para objetos lazy, min/max/mean fuerzan lectura.
    # Aquí solo mostramos metadatos seguros.
    return {
        "type": type(volume).__name__,
        "shape": volume.shape,
        "dtype": str(volume.dtype),
        "chunks": getattr(volume, "chunks", "N/A (TIF memmap)"),
    }
```

```python
# =================================================================================
# CELL 4 — NORMALIZACIÓN Y PREPROCESADO BASE (SURFACE DETECTION)
# VESUVIUS CHALLENGE — PREPROCESADO 3D PROFESIONAL
# =================================================================================

import numpy as np
import cv2
import torch
from scipy import ndimage
from typing import Tuple, Optional, Dict

# -----------------------------------------------------------------------------
# CONFIGURACIÓN GLOBAL DE PREPROCESADO
# -----------------------------------------------------------------------------
EPS = 1e-6

PREPROCESS_CFG = {
    "clip_percentiles": (0.5, 99.5),
    "zscore": True,
    "local_contrast": True,
    "local_contrast_kernel": 31,
    "gaussian_sigma": 0.5,
    "normalize_to_unit": True,
    "dtype": np.float32,
}

# -----------------------------------------------------------------------------
# CLIPPING ROBUSTO (OUTLIERS CT)
# -----------------------------------------------------------------------------
def percentile_clip(
    volume: np.ndarray,
    pmin: float,
    pmax: float
) -> np.ndarray:
    lo, hi = np.percentile(volume, [pmin, pmax])
    return np.clip(volume, lo, hi)

# -----------------------------------------------------------------------------
# NORMALIZACIÓN Z-SCORE GLOBAL
# -----------------------------------------------------------------------------
def zscore_normalize(volume: np.ndarray) -> np.ndarray:
    mean = volume.mean()
    std = volume.std()
    return (volume - mean) / (std + EPS)

# -----------------------------------------------------------------------------
# NORMALIZACIÓN A [0,1]
# -----------------------------------------------------------------------------
def unit_normalize(volume: np.ndarray) -> np.ndarray:
    vmin = volume.min()
    vmax = volume.max()
    return (volume - vmin) / (vmax - vmin + EPS)

# -----------------------------------------------------------------------------
# REALCE DE CONTRASTE LOCAL (POR SLICE)
# -----------------------------------------------------------------------------
def local_contrast_enhancement(
    volume: np.ndarray,
    kernel_size: int
) -> np.ndarray:
    out = np.empty_like(volume)
    for z in range(volume.shape[0]):
        slice_ = volume[z]
        blur = cv2.GaussianBlur(slice_, (kernel_size, kernel_size), 0)
        out[z] = slice_ - blur
    return out

# -----------------------------------------------------------------------------
# SUAVIZADO GAUSSIANO 3D (ANTI-RUIDO)
# -----------------------------------------------------------------------------
def gaussian_smooth_3d(
    volume: np.ndarray,
    sigma: float
) -> np.ndarray:
    return ndimage.gaussian_filter(volume, sigma=sigma)

# -----------------------------------------------------------------------------
# PIPELINE COMPLETO DE PREPROCESADO
# -----------------------------------------------------------------------------
def preprocess_volume_surface_detection(
    volume: np.ndarray,
    cfg: Dict = PREPROCESS_CFG
) -> np.ndarray:
    assert volume.ndim == 3, "El volumen debe ser 3D (Z, H, W)"

    v = volume.astype(cfg["dtype"], copy=False)

    # 1️⃣ Clip robusto de intensidades
    if cfg.get("clip_percentiles") is not None:
        pmin, pmax = cfg["clip_percentiles"]
        v = percentile_clip(v, pmin, pmax)

    # 2️⃣ Suavizado gaussiano leve (ruido CT)
    if cfg.get("gaussian_sigma", 0) > 0:
        v = gaussian_smooth_3d(v, cfg["gaussian_sigma"])

    # 3️⃣ Realce de contraste local (clave para superficies finas)
    if cfg.get("local_contrast", False):
        v = local_contrast_enhancement(
            v,
            kernel_size=cfg["local_contrast_kernel"]
        )

    # 4️⃣ Normalización Z-score (estabilidad entrenamiento)
    if cfg.get("zscore", False):
        v = zscore_normalize(v)

    # 5️⃣ Escalado final a [0,1]
    if cfg.get("normalize_to_unit", False):
        v = unit_normalize(v)

    return v

# -----------------------------------------------------------------------------
# CONVERSIÓN A TENSOR TORCH (CANAL PRIMERO)
# -----------------------------------------------------------------------------
def volume_to_tensor(
    volume: np.ndarray,
    add_channel: bool = True
) -> torch.Tensor:
    if add_channel:
        volume = volume[None, ...]  # (C=1, Z, H, W)
    return torch.from_numpy(volume)

# -----------------------------------------------------------------------------
# PIPELINE FINAL (LOAD → PREPROCESS → TENSOR)
# -----------------------------------------------------------------------------
def prepare_volume_for_model(
    volume: np.ndarray,
    cfg: Dict = PREPROCESS_CFG
) -> torch.Tensor:
    v = preprocess_volume_surface_detection(volume, cfg)
    t = volume_to_tensor(v)
    return t

# -----------------------------------------------------------------------------
# SANITY CHECK
# -----------------------------------------------------------------------------
def inspect_preprocessed(volume: np.ndarray) -> dict:
    return {
        "shape": volume.shape,
        "dtype": str(volume.dtype),
        "min": float(volume.min()),
        "max": float(volume.max()),
        "mean": float(volume.mean()),
        "std": float(volume.std()),
    }

# -----------------------------------------------------------------------------
# EJEMPLO DE USO (DESACTIVADO)
# -----------------------------------------------------------------------------
if False:
    vid = list(TRAIN_VOLUMES.keys())[0]
    raw = safe_load_volume(TRAIN_VOLUMES[vid], volume_id=vid)
    prep = preprocess_volume_surface_detection(raw)
    print(inspect_preprocessed(prep))
    tensor = volume_to_tensor(prep)
    print(tensor.shape)
    del raw, prep, tensor
    torch.cuda.empty_cache()
```

```python
# =================================================================================
# CELL — CONSTRUCCIÓN DE feature_volume Y TARGET (CORREGIDA)
# =================================================================================

import numpy as np
import tifffile
from pathlib import Path
from scipy.ndimage import gaussian_filter, sobel
import gc

# -----------------------------------------------------------------------------
# SELECCIÓN ROBUSTA DE VOLUMEN
# -----------------------------------------------------------------------------
# Ahora TRAIN_VOLUMES ya existe gracias a la celda anterior
assert len(TRAIN_VOLUMES) > 0, "❌ Ejecuta el PASO 1 primero para definir TRAIN_VOLUMES"

VOLUME_ID = list(TRAIN_VOLUMES.keys())[0]
VOLUME_INFO = TRAIN_VOLUMES[VOLUME_ID]
VOLUME_PATH = Path(VOLUME_INFO["tif"])

print(f"[*] Procesando Volumen ID: {VOLUME_ID}")

# -----------------------------------------------------------------------------
# CARGA DEL VOLUMEN 3D
# -----------------------------------------------------------------------------
# Leemos
raw_volume = tifffile.imread(VOLUME_PATH)

# Si es el volumen real gigante, cortamos un pedazo para no explotar la RAM (Demo mode)
# Si estás en Kaggle con 16GB RAM y el volumen es 8000x8000, ESTO ES OBLIGATORIO
if raw_volume.shape[1] > 2000:
    print("[!] Volumen muy grande detectado. Recortando centro para Feature Eng...")
    center_h, center_w = raw_volume.shape[1]//2, raw_volume.shape[2]//2
    crop_sz = 512
    raw_volume = raw_volume[:, center_h:center_h+crop_sz, center_w:center_w+crop_sz]

# Convertir a float32 y normalizar
raw_volume = raw_volume.astype(np.float32)
vmin, vmax = raw_volume.min(), raw_volume.max()
raw_volume = (raw_volume - vmin) / (vmax - vmin + 1e-6)

Z, H, W = raw_volume.shape
print(f"[*] Dimensiones de trabajo: {raw_volume.shape}")

# -----------------------------------------------------------------------------
# FEATURE VOLUME (3D → [Z, C, H, W])
# -----------------------------------------------------------------------------
C = 6
# Usamos float16 para ahorrar memoria (Features ocupan 6 veces más que el original)
feature_volume = np.zeros((Z, C, H, W), dtype=np.float16)

print("[*] Generando features slice por slice...")
for z in range(Z):
    s = raw_volume[z] # Slice 2D

    # Features básicas
    f0 = s
    f1 = gaussian_filter(s, sigma=1.0)
    f2 = gaussian_filter(s, sigma=2.0)
    
    # Gradientes (Sobel)
    gy = sobel(s, axis=0)
    gx = sobel(s, axis=1)
    f3 = gy
    f4 = gx
    f5 = np.sqrt(gx**2 + gy**2) # Magnitud del gradiente

    # Stack y guardar (convertimos a float16 al vuelo)
    stack = np.stack([f0, f1, f2, f3, f4, f5], axis=0).astype(np.float16)
    feature_volume[z] = stack

# Limpiar memoria del raw_volume ya que tenemos los features
del raw_volume, f0, f1, f2, f3, f4, f5, stack
gc.collect()

# Convertir a float32 solo si es estrictamente necesario por PyTorch después
feature_volume = np.ascontiguousarray(feature_volume).astype(np.float32)

# -----------------------------------------------------------------------------
# TARGET REAL — surface_z_index (REGRESIÓN 2D)
# -----------------------------------------------------------------------------
# Recalculamos gradiente sobre el canal de intensidad (canal 0)
# feature_volume[:, 0] es la intensidad original
print("[*] Calculando Surface Target...")

# Gradiente en Z sobre la intensidad
vol_intensity = feature_volume[:, 0, :, :]
grad_z = np.abs(np.gradient(vol_intensity, axis=0)) 

# Z de máxima energía por pixel
surface_z_index = np.argmax(grad_z, axis=0).astype(np.int64) 

# Normalizado a [0,1]
surface_z_norm = surface_z_index.astype(np.float32) / (Z - 1)

# -----------------------------------------------------------------------------
# SANITY CHECKS DUROS
# -----------------------------------------------------------------------------
assert feature_volume.shape == (Z, C, H, W)
assert surface_z_index.shape == (H, W)
assert surface_z_norm.min() >= 0.0 and surface_z_norm.max() <= 1.0
assert np.isfinite(feature_volume).all()

print(f"✅ feature_volume: {feature_volume.shape} | dtype: {feature_volume.dtype}")
print(f"✅ surface_z_norm: {surface_z_norm.shape} | dtype: {surface_z_norm.dtype}")
print(f"✅ Rango Z detectado: {surface_z_index.min()} - {surface_z_index.max()}")
```

```python
# =================================================================================
# CELL 5 — EXTRACCIÓN DE ESTRUCTURA (SKELETON ENGINE)
# VESUVIUS CHALLENGE — ENGINE GEOMÉTRICO DE PAPYRUS / SURFACE PRIORS
# =================================================================================

import numpy as np
import cv2
from skimage.filters import threshold_sauvola
from skimage.morphology import (
    skeletonize,
    remove_small_objects,
    binary_opening,
    binary_closing,
    disk,
    dilation,
)
from scipy import ndimage
from numba import jit
from typing import Tuple, List

# -----------------------------------------------------------------------------
# CONFIGURACIÓN GLOBAL DEL SKELETON ENGINE
# -----------------------------------------------------------------------------
SKELETON_CFG = {
    "sauvola_window": 25,
    "sauvola_offset": 0.05,
    "min_object_size": 50,
    "gravity_strength": 50.0,
    "gravity_blur": 3,
    "max_velocity": 3,
    "stiffness": 4.0,
    "energy_cutoff": 1000.0,
    "burn_success": 50000.0,
    "burn_fail": 400.0,
    "max_lines": 300,
}

# -----------------------------------------------------------------------------
# 1️⃣ CAMPO DE GRAVEDAD EXTREMA (PRIOR GEOMÉTRICO)
# -----------------------------------------------------------------------------
def compute_high_gravity_field(
    image_float: np.ndarray,
    cfg: dict = SKELETON_CFG,
) -> Tuple[np.ndarray, np.ndarray]:
    # Umbral dinámico para detectar fibras de papiro en CT scans ruidosos
    thresh = threshold_sauvola(
        image_float, window_size=cfg["sauvola_window"]
    )
    binary = image_float > (thresh + cfg["sauvola_offset"])
    
    # Limpieza morfológica para evitar micro-ruido
    binary = remove_small_objects(binary, min_size=cfg["min_object_size"])
    binary = binary_opening(binary, disk(1))
    binary = binary_closing(binary, disk(1))

    # Esqueleto binario (referencia topológica)
    skeleton = skeletonize(binary)

    # Cálculo del campo de gravedad basado en la distancia al esqueleto
    skel_uint8 = skeleton.astype(np.uint8)
    inverse = 1 - skel_uint8
    dist_map = cv2.distanceTransform(inverse, cv2.DIST_L2, 5)

    # La energía es proporcional al cuadrado de la distancia (atracción cuadrática)
    gravity = (dist_map ** 2) * cfg["gravity_strength"]

    if cfg["gravity_blur"] > 0:
        gravity = cv2.GaussianBlur(
            gravity,
            (cfg["gravity_blur"] * 2 + 1, cfg["gravity_blur"] * 2 + 1), # Kernel size impar
            0,
        )

    return gravity.astype(np.float32), skeleton

# -----------------------------------------------------------------------------
# 2️⃣ MOTOR FÍSICO 2DO ORDEN (DYNAMIC PROGRAMMING)
# -----------------------------------------------------------------------------
@jit(nopython=True)
def solve_flexible_dp(
    energy_map: np.ndarray,
    stiffness: float,
    max_velocity: int,
    energy_cutoff: float,
):
    h, w = energy_map.shape
    vr = 2 * max_velocity + 1

    # dp[x, y, v] almacena el costo acumulado mínimo
    dp = np.full((w, h, vr), np.inf, dtype=np.float32)
    parent_v = np.zeros((w, h, vr), dtype=np.int32)
    parent_y = np.zeros((w, h, vr), dtype=np.int32)

    # Inicialización en la primera columna
    mid_v = max_velocity
    for y in range(h):
        if energy_map[y, 0] < energy_cutoff:
            dp[0, y, mid_v] = energy_map[y, 0]

    # Transiciones: Optimización de trayectoria 2.5D
    for x in range(1, w):
        for y in range(h):
            if energy_map[y, x] > energy_cutoff:
                continue

            for v_idx in range(vr):
                v = v_idx - max_velocity
                y_prev = y - v
                
                if y_prev < 0 or y_prev >= h:
                    continue

                best_cost = np.inf
                best_prev_v = mid_v

                for pv_idx in range(vr):
                    prev_cost = dp[x - 1, y_prev, pv_idx]
                    if prev_cost >= np.inf:
                        continue

                    # Penalización por curvatura (Stiffness)
                    pv = pv_idx - max_velocity
                    curvature = v - pv
                    penalty = (curvature ** 2) * stiffness
                    cost = prev_cost + penalty

                    if cost < best_cost:
                        best_cost = cost
                        best_prev_v = pv_idx

                dp[x, y, v_idx] = best_cost + energy_map[y, x]
                parent_v[x, y, v_idx] = best_prev_v
                parent_y[x, y, v_idx] = y_prev

    return dp, parent_v, parent_y

@jit(nopython=True)
def backtrack(
    dp: np.ndarray,
    parent_v: np.ndarray,
    parent_y: np.ndarray,
    w: int,
    h: int,
    max_velocity: int,
):
    vr = 2 * max_velocity + 1
    path_y = np.zeros(w, dtype=np.int32)

    best_cost = np.inf
    cy, cv = 0, 0
    
    # Encontrar el punto final óptimo en la última columna
    for y in range(h):
        for v in range(vr):
            if dp[w - 1, y, v] < best_cost:
                best_cost = dp[w - 1, y, v]
                cy, cv = y, v

    if best_cost >= np.inf:
        return path_y # Fallback si no se encontró camino

    # Reconstrucción de la trayectoria hacia atrás
    path_y[w - 1] = cy
    for x in range(w - 1, 0, -1):
        py = parent_y[x, cy, cv]
        pv = parent_v[x, cy, cv]
        path_y[x - 1] = py
        cy, cv = py, pv

    return path_y

# -----------------------------------------------------------------------------
# 3️⃣ ORQUESTADOR DE TRACKING GEOMÉTRICO
# -----------------------------------------------------------------------------
def track_flexible_physics(
    image: np.ndarray,
    cfg: dict = SKELETON_CFG,
):
    img = image.astype(np.float32)
    if img.max() > 1.0:
        img /= 255.0

    energy_map, skeleton = compute_high_gravity_field(img, cfg)

    burn_map = np.zeros_like(energy_map, dtype=np.float32)
    h, w = energy_map.shape
    final_segments: List[np.ndarray] = []

    # Zona válida para rescatar fragmentos de papiro
    valid_zone = dilation(skeleton, disk(1))

    # Iteraciones para extraer múltiples líneas de papiro (superficies)
    for _ in range(cfg["max_lines"] * 2):
        if len(final_segments) >= cfg["max_lines"]:
            break

        current_energy = energy_map + burn_map

        dp, pv, py = solve_flexible_dp(
            current_energy,
            stiffness=cfg["stiffness"],
            max_velocity=cfg["max_velocity"],
            energy_cutoff=cfg["energy_cutoff"],
        )

        path_y = backtrack(
            dp, pv, py, w, h, cfg["max_velocity"]
        )
        
        # Validar si el camino es legítimo o ruido
        path_xy = np.column_stack((np.arange(w), path_y))
        
        # Extraer segmentos que caen en la zona del esqueleto (valid_zone)
        seg = []
        gap = 0
        for x, y in path_xy:
            if valid_zone[y, x]:
                seg.append((x, y))
                gap = 0
            else:
                gap += 1
                if gap > 5: # Tolerancia a pequeños huecos
                    if len(seg) > 5:
                        final_segments.append(np.array(seg))
                        # Quemar solo si es exitoso
                        for sx, sy in seg:
                            cv2.circle(burn_map, (sx, sy), 1, cfg["burn_success"], -1)
                    seg = []
        
        if len(seg) > 5:
            final_segments.append(np.array(seg))
            for sx, sy in seg:
                cv2.circle(burn_map, (sx, sy), 1, cfg["burn_success"], -1)
        else:
            # Si no se extrajeron segmentos, penalizar la trayectoria completa
            for x, y in path_xy:
                burn_map[y, x] += cfg["burn_fail"]

    return final_segments, skeleton

# -----------------------------------------------------------------------------
# 4️⃣ UTILIDAD: EXTRAER SKELETON 2D DESDE VOLUMEN 3D
# -----------------------------------------------------------------------------
def extract_skeleton_from_volume(
    volume: np.ndarray,
    slice_index: int = None,
):
    if slice_index is None:
        slice_index = volume.shape[0] // 2
    
    # Extraer slice y correr motor de física
    slice_img = volume[slice_index]
    segments, skeleton = track_flexible_physics(slice_img)
    
    return {
        "slice_index": slice_index,
        "segments": segments,
        "skeleton": skeleton,
    }
```

```python
# =================================================================================
# CELL 6 — SEGUIMIENTO FÍSICO DE FIBRAS (DP 2DO ORDEN)
# VESUVIUS CHALLENGE — FIBER TRACKING ENGINE (GEOMETRIC PRIOR)
# =================================================================================

import numpy as np
import cv2
from numba import jit
from scipy.ndimage import gaussian_filter
from skimage.morphology import dilation, disk
from typing import Dict, List, Tuple

# -----------------------------------------------------------------------------
# CONFIGURACIÓN GLOBAL DEL MOTOR FÍSICO
# -----------------------------------------------------------------------------
FIBER_DP_CFG: Dict = {
    "max_velocity": 3,          # rango de velocidades (curvatura permitida)
    "stiffness": 4.0,           # rigidez (penaliza cambios bruscos)
    "energy_cutoff": 1200.0,    # umbral para podar estados inviables
    "burn_success": 60000.0,    # energía para quemar trayectorias exitosas
    "burn_fail": 500.0,         # energía para penalizar fallos
    "min_segment_len": 6,       # longitud mínima de segmento válido
    "gap_tolerance": 4,         # tolerancia de gaps fuera del esqueleto
    "valid_dilation": 1,        # dilatación del esqueleto para zona válida
    "max_tracks": 300,          # máximo de fibras a rastrear
    "max_iters_factor": 5,      # factor de iteraciones por fibra
}

# -----------------------------------------------------------------------------
# UTILIDAD: NORMALIZACIÓN ROBUSTA DE MAPAS DE ENERGÍA
# -----------------------------------------------------------------------------
def normalize_energy_map(energy: np.ndarray) -> np.ndarray:
    e = energy.astype(np.float32)
    p1, p99 = np.percentile(e, (1, 99))
    e = np.clip(e, p1, p99)
    e = (e - p1) / (p99 - p1 + 1e-6)
    return e

# -----------------------------------------------------------------------------
# MOTOR DP 2DO ORDEN (OPTIMIZADO)
# -----------------------------------------------------------------------------
@jit(nopython=True)
def dp_second_order(
    energy_map: np.ndarray,
    stiffness: float,
    max_velocity: int,
    energy_cutoff: float,
):
    h, w = energy_map.shape
    vr = 2 * max_velocity + 1

    dp = np.full((w, h, vr), np.inf, dtype=np.float32)
    py = np.zeros((w, h, vr), dtype=np.int32)
    pv = np.zeros((w, h, vr), dtype=np.int32)

    mid_v = max_velocity
    for y in range(h):
        dp[0, y, mid_v] = energy_map[y, 0]

    for x in range(1, w):
        for y in range(h):
            if energy_map[y, x] > energy_cutoff:
                continue

            for v_idx in range(vr):
                v = v_idx - max_velocity
                y_prev = y - v
                if y_prev < 0 or y_prev >= h:
                    continue

                best_cost = np.inf
                best_prev_v = mid_v

                for pv_idx in range(vr):
                    prev_cost = dp[x - 1, y_prev, pv_idx]
                    if prev_cost == np.inf:
                        continue

                    pv_val = pv_idx - max_velocity
                    curvature = v - pv_val
                    penalty = (curvature * curvature) * stiffness
                    cost = prev_cost + penalty

                    if cost < best_cost:
                        best_cost = cost
                        best_prev_v = pv_idx

                dp[x, y, v_idx] = best_cost + energy_map[y, x]
                py[x, y, v_idx] = y_prev
                pv[x, y, v_idx] = best_prev_v

    return dp, py, pv

@jit(nopython=True)
def backtrack_dp(
    dp: np.ndarray,
    py: np.ndarray,
    pv: np.ndarray,
    w: int,
    h: int,
    max_velocity: int,
):
    vr = 2 * max_velocity + 1
    path_y = np.zeros(w, dtype=np.int32)

    best = np.inf
    cy, cv = 0, 0
    for y in range(h):
        for v in range(vr):
            if dp[w - 1, y, v] < best:
                best = dp[w - 1, y, v]
                cy, cv = y, v

    path_y[w - 1] = cy
    for x in range(w - 1, 0, -1):
        ny = py[x, cy, cv]
        nv = pv[x, cy, cv]
        path_y[x - 1] = ny
        cy, cv = ny, nv

    return path_y

# -----------------------------------------------------------------------------
# VALIDACIÓN TOPO-GEOMÉTRICA DE TRAYECTORIAS
# -----------------------------------------------------------------------------
def split_valid_segments(
    path_xy: np.ndarray,
    valid_zone: np.ndarray,
    min_len: int,
    gap_tol: int,
) -> List[np.ndarray]:
    segments: List[np.ndarray] = []
    seg: List[Tuple[int, int]] = []
    gap = 0

    for x, y in path_xy:
        if valid_zone[y, x]:
            if gap > gap_tol:
                if len(seg) >= min_len:
                    segments.append(np.asarray(seg, dtype=np.int32))
                seg = []
            gap = 0
            seg.append((x, y))
        else:
            gap += 1

    if len(seg) >= min_len:
        segments.append(np.asarray(seg, dtype=np.int32))

    return segments

# -----------------------------------------------------------------------------
# ORQUESTADOR PRINCIPAL DE SEGUIMIENTO DE FIBRAS
# -----------------------------------------------------------------------------
def track_fibers_dp(
    energy_map: np.ndarray,
    skeleton: np.ndarray,
    cfg: Dict = FIBER_DP_CFG,
):
    h, w = energy_map.shape

    energy = normalize_energy_map(energy_map)
    energy = gaussian_filter(energy, sigma=0.5)

    burn_map = np.zeros_like(energy, dtype=np.float32)
    fibers: List[np.ndarray] = []

    valid_zone = dilation(skeleton, disk(cfg["valid_dilation"]))

    max_iters = cfg["max_tracks"] * cfg["max_iters_factor"]

    for _ in range(max_iters):
        if len(fibers) >= cfg["max_tracks"]:
            break

        current_energy = energy + burn_map

        dp, py, pv = dp_second_order(
            current_energy,
            stiffness=cfg["stiffness"],
            max_velocity=cfg["max_velocity"],
            energy_cutoff=cfg["energy_cutoff"],
        )

        path_y = backtrack_dp(
            dp, py, pv, w, h, cfg["max_velocity"]
        )
        path_xy = np.column_stack((np.arange(w), path_y))

        segments = split_valid_segments(
            path_xy,
            valid_zone,
            min_len=cfg["min_segment_len"],
            gap_tol=cfg["gap_tolerance"],
        )

        if not segments:
            for x, y in path_xy:
                if 0 <= y < h:
                    burn_map[y, x] += cfg["burn_fail"]
            continue

        for seg in segments:
            fibers.append(seg)
            for x, y in seg:
                cv2.circle(
                    burn_map,
                    (int(x), int(y)),
                    1,
                    cfg["burn_success"],
                    -1,
                )

    return fibers

# -----------------------------------------------------------------------------
# INTERFAZ DE ALTO NIVEL (IMAGEN 2D)
# -----------------------------------------------------------------------------
def run_fiber_tracking(
    image_2d: np.ndarray,
    gravity_map: np.ndarray,
    skeleton: np.ndarray,
    cfg: Dict = FIBER_DP_CFG,
):
    assert image_2d.ndim == 2
    fibers = track_fibers_dp(
        energy_map=gravity_map,
        skeleton=skeleton,
        cfg=cfg,
    )
    return {
        "fibers": fibers,
        "count": len(fibers),
        "skeleton": skeleton,
    }

# -----------------------------------------------------------------------------
# INTERFAZ DE ALTO NIVEL (VOLUMEN 3D)
# -----------------------------------------------------------------------------
def run_fiber_tracking_volume(
    volume: np.ndarray,
    gravity_volume: np.ndarray,
    skeleton_volume: np.ndarray,
    slice_index: int,
    cfg: Dict = FIBER_DP_CFG,
):
    img = volume[slice_index]
    grav = gravity_volume[slice_index]
    skel = skeleton_volume[slice_index]

    return run_fiber_tracking(img, grav, skel, cfg)

# -----------------------------------------------------------------------------
# SANITY CHECK (DESACTIVADO)
# -----------------------------------------------------------------------------
if False:
    sl = 10
    out = run_fiber_tracking(
        image_2d=vol[sl],
        gravity_map=grav_vol[sl],
        skeleton=skel_vol[sl],
    )
    print(out["count"])
```

```python
# =================================================================================
# CELL 7 — ORQUESTADOR MULTI-TRAYECTORIA
# VESUVIUS CHALLENGE — MULTI-FIBER / MULTI-SURFACE COORDINATOR
# =================================================================================

import numpy as np
import cv2
from typing import Dict, List
from scipy.ndimage import gaussian_filter
from skimage.morphology import dilation, disk

# -----------------------------------------------------------------------------
# CONFIGURACIÓN GLOBAL DEL ORQUESTADOR
# -----------------------------------------------------------------------------
ORCHESTRATOR_CFG: Dict = {
    "max_slices": None,          # None = todas las slices
    "slice_stride": 1,           # stride en Z
    "min_fibers_per_slice": 5,   # descartar slices pobres
    "merge_radius_xy": 2.5,      # radio para merge inter-slice
    "merge_radius_z": 1.5,       # tolerancia en Z
    "min_track_length": 15,      # longitud mínima 3D
    "smoothing_sigma": 1.0,      # suavizado post-merge
}

# -----------------------------------------------------------------------------
# UTILIDADES GEOMÉTRICAS
# -----------------------------------------------------------------------------
def polyline_length_2d(poly: np.ndarray) -> float:
    if len(poly) < 2:
        return 0.0
    d = np.diff(poly.astype(np.float32), axis=0)
    return float(np.sum(np.linalg.norm(d, axis=1)))

def polyline_length_3d(poly: np.ndarray) -> float:
    if len(poly) < 2:
        return 0.0
    d = np.diff(poly.astype(np.float32), axis=0)
    return float(np.sum(np.linalg.norm(d, axis=1)))

def smooth_polyline(poly: np.ndarray, sigma: float) -> np.ndarray:
    if len(poly) < 3:
        return poly
    out = np.zeros_like(poly, dtype=np.float32)
    for i in range(poly.shape[1]):
        out[:, i] = gaussian_filter(poly[:, i], sigma=sigma)
    return out

# -----------------------------------------------------------------------------
# MERGE DE TRAYECTORIAS ENTRE SLICES (ENSAMBLE 3D)
# -----------------------------------------------------------------------------
def merge_fibers_across_slices(
    fibers_by_slice: Dict[int, List[np.ndarray]],
    cfg: Dict,
):
    tracks_3d: List[np.ndarray] = []

    sorted_slices = sorted(fibers_by_slice.keys())
    active_tracks: List[np.ndarray] = []

    for z in sorted_slices:
        slice_fibers = fibers_by_slice[z]

        used = [False] * len(slice_fibers)

        new_active_tracks: List[np.ndarray] = []

        for track in active_tracks:
            last_pt = track[-1]
            best_idx = -1
            best_dist = np.inf

            for i, fib in enumerate(slice_fibers):
                if used[i]:
                    continue
                first_pt = fib[0]
                dz = abs(first_pt[2] - last_pt[2])
                if dz > cfg["merge_radius_z"]:
                    continue

                dxy = np.linalg.norm(first_pt[:2] - last_pt[:2])
                if dxy < best_dist and dxy <= cfg["merge_radius_xy"]:
                    best_dist = dxy
                    best_idx = i

            if best_idx >= 0:
                used[best_idx] = True
                merged = np.vstack([track, slice_fibers[best_idx]])
                new_active_tracks.append(merged)
            else:
                tracks_3d.append(track)

        for i, fib in enumerate(slice_fibers):
            if not used[i]:
                new_active_tracks.append(fib)

        active_tracks = new_active_tracks

    tracks_3d.extend(active_tracks)
    return tracks_3d

# -----------------------------------------------------------------------------
# ORQUESTADOR PRINCIPAL MULTI-SLICE
# -----------------------------------------------------------------------------
def orchestrate_multi_trajectory(
    volume: np.ndarray,
    gravity_volume: np.ndarray,
    skeleton_volume: np.ndarray,
    fiber_cfg: Dict,
    orch_cfg: Dict = ORCHESTRATOR_CFG,
):
    z_slices = volume.shape[0]
    max_slices = orch_cfg["max_slices"] or z_slices

    fibers_by_slice: Dict[int, List[np.ndarray]] = {}

    for z in range(0, min(z_slices, max_slices), orch_cfg["slice_stride"]):
        gravity = gravity_volume[z]
        skeleton = skeleton_volume[z]

        if skeleton.sum() < 10:
            continue

        fibers_2d = track_fibers_dp(
            energy_map=gravity,
            skeleton=skeleton,
            cfg=fiber_cfg,
        )

        if len(fibers_2d) < orch_cfg["min_fibers_per_slice"]:
            continue

        fibers_3d = []
        for fib in fibers_2d:
            z_col = np.full((fib.shape[0], 1), z, dtype=np.int32)
            fibers_3d.append(
                np.concatenate([fib.astype(np.int32), z_col], axis=1)
            )

        fibers_by_slice[z] = fibers_3d

    # Merge inter-slice
    tracks_3d = merge_fibers_across_slices(
        fibers_by_slice,
        cfg=orch_cfg,
    )

    # Filtrado por longitud mínima
    final_tracks: List[np.ndarray] = []
    for tr in tracks_3d:
        if polyline_length_3d(tr) >= orch_cfg["min_track_length"]:
            final_tracks.append(tr)

    # Suavizado final
    smoothed_tracks = [
        smooth_polyline(tr, orch_cfg["smoothing_sigma"])
        for tr in final_tracks
    ]

    return {
        "tracks_3d": smoothed_tracks,
        "num_tracks": len(smoothed_tracks),
        "slices_used": len(fibers_by_slice),
    }

# -----------------------------------------------------------------------------
# SANITY CHECK (DESACTIVADO)
# -----------------------------------------------------------------------------
if False:
    result = orchestrate_multi_trajectory(
        volume=vol,
        gravity_volume=grav_vol,
        skeleton_volume=skel_vol,
        fiber_cfg=FIBER_DP_CFG,
    )
    print(result["num_tracks"])
```

```python
# =================================================================================
# CELL 8 — MAPAS DERIVADOS PARA DEEP LEARNING
# VESUVIUS CHALLENGE — FEATURE ENGINEERING MULTICANAL (SURFACE DETECTION)
# =================================================================================

import numpy as np
import cv2
from scipy.ndimage import (
    distance_transform_edt,
    gaussian_filter,
    sobel,
)
from skimage.filters import frangi
from skimage.morphology import dilation, disk
from typing import Dict, Tuple, List

# -----------------------------------------------------------------------------
# CONFIGURACIÓN DE FEATURES (SIN LEAKAGE DE GT)
# -----------------------------------------------------------------------------
FEATURE_CFG: Dict = {
    "use_intensity": True,
    "use_z_context": True,
    "z_window": 5,                 # contexto 2.5D
    "use_distance_to_skeleton": True,
    "use_gravity_field": True,     # campo de gravedad del Skeleton Engine
    "use_orientation": True,
    "use_curvature": True,
    "use_frangi": True,
    "frangi_sigmas": (1, 2, 3),
    "smooth_sigma": 1.0,
    "clip_percentiles": (1, 99),
}

# -----------------------------------------------------------------------------
# UTILIDADES
# -----------------------------------------------------------------------------
def normalize_robust(x: np.ndarray, pmin=1, pmax=99, eps=1e-6) -> np.ndarray:
    lo, hi = np.percentile(x, [pmin, pmax])
    x = np.clip(x, lo, hi)
    return (x - lo) / (hi - lo + eps)

def compute_orientation_map(img: np.ndarray) -> np.ndarray:
    gx = sobel(img, axis=1)
    gy = sobel(img, axis=0)
    return np.arctan2(gy, gx)

def compute_curvature_map(orientation: np.ndarray) -> np.ndarray:
    ox = sobel(orientation, axis=1)
    oy = sobel(orientation, axis=0)
    return np.sqrt(ox**2 + oy**2)

# -----------------------------------------------------------------------------
# Z-CONTEXT (2.5D STACK)
# -----------------------------------------------------------------------------
def extract_z_context(
    volume: np.ndarray,
    z: int,
    window: int,
) -> np.ndarray:
    half = window // 2
    zs = np.arange(z - half, z + half + 1)
    zs = np.clip(zs, 0, volume.shape[0] - 1)
    return volume[zs]

# -----------------------------------------------------------------------------
# FEATURE MAPS POR SLICE (SOLO INPUT FEATURES)
# -----------------------------------------------------------------------------
def build_feature_maps_for_slice(
    volume: np.ndarray,
    skeleton_slice: np.ndarray,
    gravity_slice: np.ndarray,
    z: int,
    cfg: Dict = FEATURE_CFG,
) -> Tuple[np.ndarray, Dict[str, int]]:

    feats: List[np.ndarray] = []
    channel_index: Dict[str, int] = {}
    ch = 0

    # -------------------------------
    # INTENSIDAD BASE
    # -------------------------------
    if cfg["use_intensity"]:
        img = volume[z].astype(np.float32)
        img = gaussian_filter(img, cfg["smooth_sigma"])
        img = normalize_robust(img, *cfg["clip_percentiles"])
        feats.append(img)
        channel_index["intensity"] = ch
        ch += 1

    # -------------------------------
    # CONTEXTO Z (2.5D)
    # -------------------------------
    if cfg["use_z_context"]:
        z_stack = extract_z_context(volume, z, cfg["z_window"]).astype(np.float32)
        z_stack = gaussian_filter(z_stack, cfg["smooth_sigma"])
        z_stack = normalize_robust(z_stack, *cfg["clip_percentiles"])
        for i in range(z_stack.shape[0]):
            feats.append(z_stack[i])
            channel_index[f"z_ctx_{i}"] = ch
            ch += 1

    # -------------------------------
    # DISTANCIA AL ESQUELETO
    # -------------------------------
    if cfg["use_distance_to_skeleton"]:
        sk = skeleton_slice.astype(bool)
        inv = ~sk
        dist = distance_transform_edt(inv)
        dist = gaussian_filter(dist, cfg["smooth_sigma"])
        dist = normalize_robust(dist, *cfg["clip_percentiles"])
        feats.append(dist)
        channel_index["dist_skeleton"] = ch
        ch += 1

    # -------------------------------
    # CAMPO DE GRAVEDAD (SKELETON ENGINE)
    # -------------------------------
    if cfg["use_gravity_field"]:
        g = gravity_slice.astype(np.float32)
        g = gaussian_filter(g, cfg["smooth_sigma"])
        g = normalize_robust(g, *cfg["clip_percentiles"])
        feats.append(g)
        channel_index["gravity"] = ch
        ch += 1

    # -------------------------------
    # ORIENTACIÓN LOCAL
    # -------------------------------
    if cfg["use_orientation"]:
        ori = compute_orientation_map(volume[z].astype(np.float32))
        ori = normalize_robust(ori, 1, 99)
        feats.append(ori)
        channel_index["orientation"] = ch
        ch += 1

    # -------------------------------
    # CURVATURA
    # -------------------------------
    if cfg["use_curvature"]:
        ori = compute_orientation_map(volume[z].astype(np.float32))
        curv = compute_curvature_map(ori)
        curv = gaussian_filter(curv, cfg["smooth_sigma"])
        curv = normalize_robust(curv, 1, 99)
        feats.append(curv)
        channel_index["curvature"] = ch
        ch += 1

    # -------------------------------
    # FRANGI (ESTRUCTURAS LINEALES)
    # -------------------------------
    if cfg["use_frangi"]:
        fr = frangi(
            volume[z].astype(np.float32),
            sigmas=cfg["frangi_sigmas"],
            black_ridges=False,
        )
        fr = normalize_robust(fr, 1, 99)
        feats.append(fr)
        channel_index["frangi"] = ch
        ch += 1

    feature_tensor = np.stack(feats, axis=0).astype(np.float32)
    return feature_tensor, channel_index

# -----------------------------------------------------------------------------
# FEATURE MAPS PARA TODO EL VOLUMEN
# -----------------------------------------------------------------------------
def build_feature_volume(
    volume: np.ndarray,
    skeleton_volume: np.ndarray,
    gravity_volume: np.ndarray,
    cfg: Dict = FEATURE_CFG,
):
    feature_slices = []
    channel_index = None

    for z in range(volume.shape[0]):
        feat, ch_idx = build_feature_maps_for_slice(
            volume=volume,
            skeleton_slice=skeleton_volume[z],
            gravity_slice=gravity_volume[z],
            z=z,
            cfg=cfg,
        )
        feature_slices.append(feat)
        if channel_index is None:
            channel_index = ch_idx

    feature_volume = np.stack(feature_slices, axis=0)
    return feature_volume, channel_index

```

```python
# =================================================================================
# CELL 9 — DATASET TORCH (CORREGIDO: 2.5D VOLUMETRIC PATCHES)
# VESUVIUS CHALLENGE — SURFACE DETECTION
# =================================================================================

import torch
from torch.utils.data import Dataset
import numpy as np
import albumentations as A
from albumentations.pytorch import ToTensorV2

# -----------------------------------------------------------------------------
# DATASET: PARCHES VOLUMÉTRICOS CENTRADOS EN LA SUPERFICIE
# -----------------------------------------------------------------------------
class VesuviusSurfaceDataset(Dataset):
    """
    Genera parches 3D centrados en la profundidad estimada del papiro.
    
    Input (Lazy Arrays):
        feature_volume   : Zarr/Array [Z, C, H, W]
        surface_mask     : Zarr/Array [Z, H, W] (Ground Truth Binario)
        surface_z_map    : Array [H, W] (Mapa de altura aproximado para centrar el crop)

    Output (Tensor):
        image  : [Z_window, C, Patch_H, Patch_W]
        mask   : [Z_window, Patch_H, Patch_W] (o solo la central)
    """

    def __init__(
        self,
        feature_volume,          # [Z, C, H, W]
        surface_mask,            # [Z, H, W]
        surface_z_map,           # [H, W] - Usado para encontrar el centro Z
        patch_size: int = 256,
        z_window: int = 16,      # Debe coincidir con NUM_SLICES_CONTEXT
        stride: int = 128,
        train: bool = True,
        transform: A.Compose = None,
    ):
        # Validaciones de integridad
        assert z_window % 2 == 0, "z_window debe ser par para centrar correctamente"
        
        self.volume = feature_volume
        self.mask = surface_mask
        self.z_map = surface_z_map
        
        self.patch_size = patch_size
        self.z_window = z_window
        self.half_z = z_window // 2
        self.stride = stride
        self.train = train

        # Dimensiones globales
        self.Z = feature_volume.shape[0]
        self.H = feature_volume.shape[2] # [Z, C, H, W]
        self.W = feature_volume.shape[3]

        # Definir transformaciones base si no se proveen
        if transform is None and train:
            self.transform = A.Compose([
                A.HorizontalFlip(p=0.5),
                A.VerticalFlip(p=0.5),
                A.RandomRotate90(p=0.5),
                # A.Transpose(p=0.5),
            ])
        else:
            self.transform = transform

        # Generar coordenadas de los parches válidos
        self.coords = self._build_valid_coords()
        print(f"[*] Dataset creado. Patches válidos: {len(self.coords)} | Train: {train}")

    def _build_valid_coords(self):
        """
        Genera una lista de coordenadas (y, x) que contienen superficie válida.
        Evita generar parches vacíos (fondo negro) para no desperdiciar cómputo.
        """
        coords = []
        # Iteramos con stride espacial
        for y in range(0, self.H - self.patch_size + 1, self.stride):
            for x in range(0, self.W - self.patch_size + 1, self.stride):
                
                # Verificamos si hay papiro en este parche usando el z_map
                # Si el z_map es 0 o muy bajo, asumimos que no hay superficie
                patch_z_map = self.z_map[y : y + self.patch_size, x : x + self.patch_size]
                
                # Criterio simple: Si hay información de altura válida en el parche
                if np.count_nonzero(patch_z_map) > (self.patch_size**2 * 0.05): # >5% válido
                    coords.append((y, x))
                    
        return coords

    def __len__(self):
        return len(self.coords)

    def __getitem__(self, idx):
        y, x = self.coords[idx]
        
        # 1. Determinar el centro Z del parche
        # Usamos la mediana del mapa de altura en este parche para ser robustos a outliers
        patch_z_values = self.z_map[y : y + self.patch_size, x : x + self.patch_size]
        valid_z = patch_z_values[patch_z_values > 0]
        
        if len(valid_z) > 0:
            center_z = int(np.median(valid_z))
        else:
            center_z = self.Z // 2 # Fallback
            
        # 2. Calcular límites Z (Clamping para no salir del volumen)
        z_start = center_z - self.half_z
        z_end = center_z + self.half_z
        
        # Ajuste de bordes Z
        if z_start < 0:
            z_start = 0
            z_end = self.z_window
        elif z_end > self.Z:
            z_end = self.Z
            z_start = self.Z - self.z_window
            
        # 3. Extracción de Datos (LAZY SLICING)
        # Importante: feature_volume es [Z, C, H, W]
        # Extraemos: [z_window, C, patch_size, patch_size]
        features = self.volume[z_start:z_end, :, y:y+self.patch_size, x:x+self.patch_size]
        
        # Ground Truth Mask [z_window, patch_size, patch_size]
        mask_patch = self.mask[z_start:z_end, y:y+self.patch_size, x:x+self.patch_size]

        # 4. Convertir a Numpy (Float32) en memoria local
        features = features.astype(np.float32)
        mask_patch = mask_patch.astype(np.float32)

        # 5. Data Augmentation (Sincronizada Espacialmente)
        if self.transform:
            # Albumentations espera [H, W, C], así que tenemos que hacer trucos
            # Concatenamos features y máscara en el eje de canales para rotarlos juntos
            # features: [Z, C, H, W] -> reshape -> [H, W, Z*C]
            D, C, H, W = features.shape
            
            # Aplanar Z y C para Albumentations
            feat_flat = features.transpose(2, 3, 0, 1).reshape(H, W, -1) 
            mask_flat = mask_patch.transpose(1, 2, 0) # [H, W, Z]
            
            # Aplicar transformación
            augmented = self.transform(image=feat_flat, mask=mask_flat)
            feat_aug = augmented['image']
            mask_aug = augmented['mask']
            
            # Restaurar dimensiones originales
            # [H, W, Z*C] -> [Z, C, H, W]
            features = feat_aug.reshape(H, W, D, C).transpose(2, 3, 0, 1)
            # [H, W, Z] -> [Z, H, W]
            mask_patch = mask_aug.transpose(2, 0, 1)

        # 6. Preparar Tensors PyTorch
        # Input Model: [Z, C, H, W]
        img_tensor = torch.from_numpy(np.ascontiguousarray(features))
        
        # Target Model: [1, H, W] (Usamos la rebanada central como objetivo principal)
        # Ojo: Si usas CompositeLoss 3D, devuelve todo el volumen de máscara.
        # Aquí devolvemos la máscara central para 2D supervision.
        mid_idx = self.half_z
        mask_tensor = torch.from_numpy(np.ascontiguousarray(mask_patch[mid_idx])).unsqueeze(0) # [1, H, W]
        
        # También devolvemos el esqueleto del target para la SkeletonLoss
        # (Se calcula al vuelo para ahorrar espacio en disco)
        from skimage.morphology import skeletonize
        skel_np = skeletonize(mask_patch[mid_idx] > 0.5).astype(np.float32)
        skel_tensor = torch.from_numpy(np.ascontiguousarray(skel_np)).unsqueeze(0)

        return {
            "image": img_tensor,   # [Z, C, H, W] -> Listo para tu VesuviusSurfaceNet 2.5D
            "mask": mask_tensor,   # [1, H, W]
            "skel": skel_tensor    # [1, H, W]
        }

# -----------------------------------------------------------------------------
# COLLATE FN (CORREGIDO PARA DICCIONARIOS)
# -----------------------------------------------------------------------------
def vesuvius_collate_fn(batch):
    return {
        "image": torch.stack([b["image"] for b in batch], dim=0), # [B, Z, C, H, W]
        "mask": torch.stack([b["mask"] for b in batch], dim=0),   # [B, 1, H, W]
        "skel": torch.stack([b["skel"] for b in batch], dim=0),   # [B, 1, H, W]
    }
```

```python
# =================================================================================
# CELL 10A — MODELO 2.5D: SURFACE SEGMENTATION (CORREGIDO & COMPLETO)
# VESUVIUS CHALLENGE — FASE OFICIAL
# =================================================================================

import torch
import torch.nn as nn
import torch.nn.functional as F

# -----------------------------------------------------------------------------
# BLOQUES BÁSICOS 2D
# -----------------------------------------------------------------------------
class ConvBNAct(nn.Module):
    def __init__(self, in_ch, out_ch, k=3, s=1, p=1, act=True):
        super().__init__()
        self.conv = nn.Conv2d(in_ch, out_ch, k, s, p, bias=False)
        self.bn = nn.BatchNorm2d(out_ch)
        self.act = nn.SiLU(inplace=True) if act else nn.Identity()

    def forward(self, x):
        return self.act(self.bn(self.conv(x)))

class ResidualBlock2D(nn.Module):
    def __init__(self, ch):
        super().__init__()
        self.conv1 = ConvBNAct(ch, ch)
        self.conv2 = ConvBNAct(ch, ch, act=False)
        self.act = nn.SiLU(inplace=True)

    def forward(self, x):
        return self.act(x + self.conv2(self.conv1(x)))

# -----------------------------------------------------------------------------
# ENCODER 2D (Soporte Multicanal Dinámico)
# -----------------------------------------------------------------------------
class Encoder2D(nn.Module):
    def __init__(self, in_ch):
        super().__init__()
        # Stem dinámico: acepta 'in_ch' (ej. 15) en lugar de fijo a 1
        self.stem = ConvBNAct(in_ch, 64)

        self.enc1 = nn.Sequential(ConvBNAct(64, 64), ResidualBlock2D(64))
        self.enc2 = nn.Sequential(ConvBNAct(64, 128, s=2), ResidualBlock2D(128))
        self.enc3 = nn.Sequential(ConvBNAct(128, 256, s=2), ResidualBlock2D(256))
        self.enc4 = nn.Sequential(ConvBNAct(256, 384, s=2), ResidualBlock2D(384))

    def forward(self, x):
        x0 = self.stem(x)
        x1 = self.enc1(x0)
        x2 = self.enc2(x1)
        x3 = self.enc3(x2)
        x4 = self.enc4(x3)
        return x1, x2, x3, x4

# -----------------------------------------------------------------------------
# Z-CONTEXT BLOCK (La "Magia" 2.5D)
# -----------------------------------------------------------------------------
class ZContextBlock(nn.Module):
    """
    Procesa la secuencia latente en el eje Z usando una GRU.
    Input: [B*Z, C, H, W] -> Output: [B*Z, C, H, W] (Enriquecido)
    """
    def __init__(self, channels, z_dim):
        super().__init__()
        self.z_dim = z_dim
        self.channels = channels
        
        # Reducimos canales para la GRU (eficiencia)
        self.gru_dim = channels // 2
        
        self.gru = nn.GRU(
            input_size=channels,
            hidden_size=self.gru_dim,
            num_layers=1,
            batch_first=True,
            bidirectional=True
        )
        
        # Proyección de vuelta al espacio de canales original
        self.proj = nn.Sequential(
            nn.Linear(self.gru_dim * 2, channels),
            nn.Sigmoid() # Mecanismo de atención/gating
        )

    def forward(self, x):
        # x shape: [B*Z, C, H, W]
        BZ, C, H, W = x.shape
        B = BZ // self.z_dim
        Z = self.z_dim

        # 1. Global Average Pooling por slice [B*Z, C]
        gap = F.adaptive_avg_pool2d(x, 1).view(B, Z, C)
        
        # 2. Modelado de Secuencia Z (GRU)
        # La GRU aprende la continuidad del papiro entre slices
        gru_out, _ = self.gru(gap) # [B, Z, C] (Bidireccional concat)
        
        # 3. Calcular pesos de atención
        weights = self.proj(gru_out) # [B, Z, C]
        
        # 4. Aplicar atención al volumen original
        weights = weights.view(BZ, C, 1, 1)
        
        return x * weights + x # Residual connection

# -----------------------------------------------------------------------------
# DECODER 2D
# -----------------------------------------------------------------------------
class Decoder2D(nn.Module):
    def __init__(self):
        super().__init__()
        self.up3 = ConvBNAct(384 + 256, 256)
        self.up2 = ConvBNAct(256 + 128, 128)
        self.up1 = ConvBNAct(128 + 64, 64)

    def forward(self, x1, x2, x3, x4):
        d3 = F.interpolate(x4, size=x3.shape[-2:], mode="bilinear", align_corners=False)
        d3 = self.up3(torch.cat([d3, x3], dim=1))

        d2 = F.interpolate(d3, size=x2.shape[-2:], mode="bilinear", align_corners=False)
        d2 = self.up2(torch.cat([d2, x2], dim=1))

        d1 = F.interpolate(d2, size=x1.shape[-2:], mode="bilinear", align_corners=False)
        d1 = self.up1(torch.cat([d1, x1], dim=1))
        return d1

# -----------------------------------------------------------------------------
# MODELO FINAL 2.5D (ORQUESTADOR)
# -----------------------------------------------------------------------------
class VesuviusSurfaceNet(nn.Module):
    def __init__(self, in_channels, z_dim=16):
        super().__init__()
        self.z_dim = z_dim
        
        # 1. Encoder (Multicanal)
        self.encoder = Encoder2D(in_ch=in_channels)
        
        # 2. Contexto Volumétrico (En el bottleneck)
        self.z_context = ZContextBlock(channels=384, z_dim=z_dim)
        
        # 3. Decoder
        self.decoder = Decoder2D()
        
        # 4. Heads (Dual Output para Training Loop)
        self.head = nn.Conv2d(64, 1, kernel_size=1)
        self.skeleton_head = nn.Conv2d(64, 1, kernel_size=1) # Auxiliar

    def forward(self, x):
        """
        Input: [B, Z, C, H, W]
        Output: (logits_center, skel_center)
        """
        if x.ndim == 5:
            B, Z, C, H, W = x.shape
            
            # A. Distributed Encoding: Colapsar B y Z
            x = x.view(B * Z, C, H, W)
            
            # B. Feature Extraction
            x1, x2, x3, x4 = self.encoder(x) # x4 es [B*Z, 384, H/8, W/8]
            
            # C. Z-Context Awareness (Bottleneck)
            x4 = self.z_context(x4)
            
            # D. Decoding
            d = self.decoder(x1, x2, x3, x4) # [B*Z, 64, H, W]
            
            # E. Heads
            logits = self.head(d)       # [B*Z, 1, H, W]
            skel = self.skeleton_head(d)
            
            # F. Reshape & Select Center Slice
            # Solo nos importa la predicción de la rebanada central del contexto
            logits = logits.view(B, Z, 1, H, W)
            skel = skel.view(B, Z, 1, H, W)
            
            mid = Z // 2
            return logits[:, mid], skel[:, mid]
            
        else:
            # Fallback seguro para inferencia 2D simple (si fuera necesario)
            x1, x2, x3, x4 = self.encoder(x)
            d = self.decoder(x1, x2, x3, x4)
            return self.head(d), self.skeleton_head(d)
```

```python
# =================================================================================
# CELL 11 — EMA (Exponential Moving Average) & OPTIMIZER FACTORY
# VESUVIUS CHALLENGE — SURFACE DETECTION
# =================================================================================

import torch
import torch.optim as optim
from copy import deepcopy
from typing import Optional, Dict, Any
from dataclasses import dataclass

# -----------------------------------------------------------------------------
# CLASE EMA (OPTIMIZADA PARA VRAM)
# -----------------------------------------------------------------------------
class ModelEMA:
    """
    Exponential Moving Average (EMA) para modelos PyTorch.
    Optimizado para Kaggle: Realiza backup en CPU para ahorrar VRAM durante validación.
    """

    def __init__(
        self,
        model: torch.nn.Module,
        decay: float = 0.999,
        warmup_steps: int = 0,
        device: Optional[torch.device] = None,
    ):
        self.decay = decay
        self.warmup_steps = warmup_steps
        self.model = model
        
        # Creamos la sombra (EMA) y congelamos gradientes
        self.ema = deepcopy(model).eval()
        for p in self.ema.parameters():
            p.requires_grad_(False)

        if device is not None:
            self.ema.to(device=device)
            
        self.num_updates = 0
        self.backup = None # Almacenará state_dict en CPU

    @torch.no_grad()
    def _get_decay(self) -> float:
        if self.warmup_steps > 0:
            return min(self.decay, (self.num_updates + 1) / self.warmup_steps)
        return self.decay

    @torch.no_grad()
    def update(self, model: Optional[torch.nn.Module] = None):
        """Actualiza los pesos del modelo EMA con suavizado exponencial."""
        self.num_updates += 1
        d = self._get_decay()
        
        actual_model = model if model is not None else self.model
        msd = actual_model.state_dict()
        esd = self.ema.state_dict()

        for k, v in esd.items():
            if v.dtype.is_floating_point:
                v.mul_(d).add_(msd[k].detach(), alpha=1.0 - d)
            else:
                v.copy_(msd[k])

    def apply_shadow(self):
        """
        Intercambia pesos: Guarda el modelo actual en CPU y carga el EMA en GPU.
        Útil para validar con los pesos promediados sin perder el estado actual.
        """
        # IMPORTANTE: Backup a CPU para no explotar la VRAM de la GPU
        self.backup = {k: v.cpu().clone() for k, v in self.model.state_dict().items()}
        self.model.load_state_dict(self.ema.state_dict())
    
    def restore(self):
        """Restaura los pesos originales de entrenamiento."""
        if self.backup is not None:
            self.model.load_state_dict(self.backup)
            self.backup = None

    def state_dict(self) -> Dict:
        return {
            "ema": self.ema.state_dict(),
            "num_updates": self.num_updates,
        }

    def load_state_dict(self, state: Dict):
        self.ema.load_state_dict(state["ema"])
        self.num_updates = state.get("num_updates", 0)

# -----------------------------------------------------------------------------
# CONTEXT MANAGER (SINTAXIS PYTHONICA)
# -----------------------------------------------------------------------------
class EMAContext:
    """
    Uso:
        with EMAContext(ema):
            val_loss = validate(model, loader)
    """
    def __init__(self, ema: Optional[ModelEMA]):
        self.ema = ema

    def __enter__(self):
        if self.ema is not None:
            self.ema.apply_shadow()

    def __exit__(self, exc_type, exc_val, exc_tb):
        if self.ema is not None:
            self.ema.restore()

# -----------------------------------------------------------------------------
# CONFIGURACIÓN Y BUILDERS (FACTORIES)
# -----------------------------------------------------------------------------
@dataclass
class OptimizerConfig:
    lr: float = 2e-4
    weight_decay: float = 1e-2
    eta_min: float = 1e-6
    epochs: int = 25

def build_optimizer(model: torch.nn.Module, cfg: OptimizerConfig):
    """Construye AdamW, el estándar para Transformers/ResNets modernos."""
    return optim.AdamW(
        model.parameters(), 
        lr=cfg.lr, 
        weight_decay=cfg.weight_decay
    )

def build_scheduler(optimizer: torch.optim.Optimizer, cfg: OptimizerConfig):
    """Cosine Annealing para convergencia suave."""
    return optim.lr_scheduler.CosineAnnealingLR(
        optimizer,
        T_max=cfg.epochs,
        eta_min=cfg.eta_min,
    )
```

```python
# =================================================================================
# CELL 12 — LOSS FUNCTIONS AVANZADAS (DUAL HEAD SUPPORT)
# VESUVIUS CHALLENGE — SURFACE DETECTION
# =================================================================================

import torch
import torch.nn as nn
import torch.nn.functional as F
from typing import Dict, Optional, Tuple, Union

# -----------------------------------------------------------------------------
# 1. BLOQUES BÁSICOS (DICE, FOCAL)
# -----------------------------------------------------------------------------
class DiceLoss(nn.Module):
    def __init__(self, smooth: float = 1.0, mode: str = "train"):
        super().__init__()
        self.smooth = smooth
        self.mode = mode

    def forward(self, logits: torch.Tensor, targets: torch.Tensor) -> torch.Tensor:
        # Sigmoid para convertir logits a probabilidades [0, 1]
        probs = torch.sigmoid(logits)
        
        # Aplanar tensores para cálculo global en el batch (o por imagen)
        probs = probs.view(probs.size(0), -1)
        targets = targets.view(targets.size(0), -1)

        intersection = (probs * targets).sum(dim=1)
        union = probs.sum(dim=1) + targets.sum(dim=1)

        # Dice Coeff: (2*I + S) / (U + S)
        dice = (2.0 * intersection + self.smooth) / (union + self.smooth)
        
        return 1.0 - dice.mean()

class FocalLoss(nn.Module):
    def __init__(self, alpha: float = 0.25, gamma: float = 2.0):
        super().__init__()
        self.alpha = alpha
        self.gamma = gamma

    def forward(self, logits: torch.Tensor, targets: torch.Tensor) -> torch.Tensor:
        bce_loss = F.binary_cross_entropy_with_logits(
            logits, targets, reduction="none"
        )
        pt = torch.exp(-bce_loss) # prob of ground truth class
        focal_loss = self.alpha * (1 - pt) ** self.gamma * bce_loss
        return focal_loss.mean()

# -----------------------------------------------------------------------------
# 2. LOSS COMPUESTA (SURFACE + SKELETON AUXILIARY)
# -----------------------------------------------------------------------------
class CompositeSurfaceLoss(nn.Module):
    """
    Loss maestra que maneja:
    1. Pérdida principal (Surface): BCE + Dice + Focal
    2. Pérdida auxiliar (Skeleton): Dice (para forzar continuidad)
    """

    def __init__(
        self,
        w_bce: float = 0.3,
        w_dice: float = 0.4,
        w_focal: float = 0.3,
        w_aux_skel: float = 0.2, # Peso para la cabeza auxiliar
    ):
        super().__init__()
        self.bce = nn.BCEWithLogitsLoss()
        self.dice = DiceLoss(smooth=1.0)
        self.focal = FocalLoss(alpha=0.25, gamma=2.0)

        # Pesos dinámicos
        self.weights = {
            "bce": w_bce,
            "dice": w_dice,
            "focal": w_focal,
            "aux": w_aux_skel
        }

    def update_weights(self, new_weights: Dict[str, float]):
        """Permite Curriculum Learning (cambiar pesos durante el entrenamiento)"""
        for k, v in new_weights.items():
            if k in self.weights:
                self.weights[k] = v

    def forward(
        self, 
        outputs: Union[torch.Tensor, Tuple[torch.Tensor, torch.Tensor]], 
        target_surface: torch.Tensor, 
        target_skeleton: Optional[torch.Tensor] = None
    ) -> torch.Tensor:
        """
        outputs: Puede ser tensor único (logits) o tupla (logits, skel_logits)
        target_surface: Máscara Ground Truth [B, 1, H, W]
        target_skeleton: Esqueleto Ground Truth [B, 1, H, W]
        """
        
        # 1. Desempaquetado seguro (Soporte Dual-Head de Cell 10A)
        if isinstance(outputs, (tuple, list)):
            logits_surf, logits_skel = outputs
        else:
            logits_surf, logits_skel = outputs, None

        # 2. Pérdida Principal (Surface Segmentation)
        loss_bce = self.bce(logits_surf, target_surface)
        loss_dice = self.dice(logits_surf, target_surface)
        loss_focal = self.focal(logits_surf, target_surface)

        total_loss = (
            self.weights["bce"] * loss_bce +
            self.weights["dice"] * loss_dice +
            self.weights["focal"] * loss_focal
        )

        # 3. Pérdida Auxiliar (Skeleton Head)
        # Solo calculamos si el modelo devolvió skel_logits y tenemos targets
        if logits_skel is not None and target_skeleton is not None:
            # Usamos DiceLoss para el esqueleto (maneja bien el desbalance extremo)
            loss_aux = self.dice(logits_skel, target_skeleton)
            total_loss += self.weights["aux"] * loss_aux

        return total_loss

# -----------------------------------------------------------------------------
# 3. LOSS CURRICULUM ENGINE
# -----------------------------------------------------------------------------
class LossCurriculum:
    """
    Ajusta la estrategia de Loss según la época.
    Ejemplo: Empezar con BCE fuerte, terminar refinando con Dice.
    """
    def __init__(self, criterion: CompositeSurfaceLoss, schedule: list):
        self.criterion = criterion
        # Schedule formato: [(epoch, {dict_pesos}), ...]
        self.schedule = sorted(schedule, key=lambda x: x[0])

    def step(self, epoch: int):
        for e, weights in reversed(self.schedule):
            if epoch >= e:
                self.criterion.update_weights(weights)
                break
```

```python
# =================================================================================
# CELL 13 — OPTIMIZER & SCHEDULER
# VESUVIUS CHALLENGE — SURFACE DETECTION
# =================================================================================

import math
import torch
from torch.optim import AdamW
from torch.optim.lr_scheduler import _LRScheduler


# -----------------------------------------------------------------------------
# OPTIMIZER FACTORY
# -----------------------------------------------------------------------------
def build_optimizer(model, cfg):
    """
    Optimizer profesional con weight decay desacoplado.
    """
    no_decay = ["bias", "LayerNorm.weight", "layer_norm.weight"]

    param_groups = [
        {
            "params": [
                p for n, p in model.named_parameters()
                if not any(nd in n for nd in no_decay) and p.requires_grad
            ],
            "weight_decay": cfg.weight_decay,
        },
        {
            "params": [
                p for n, p in model.named_parameters()
                if any(nd in n for nd in no_decay) and p.requires_grad
            ],
            "weight_decay": 0.0,
        },
    ]

    optimizer = AdamW(
        param_groups,
        lr=cfg.lr,
        betas=cfg.betas,
        eps=cfg.eps,
    )
    return optimizer


# -----------------------------------------------------------------------------
# COSINE WARMUP SCHEDULER (CUSTOM)
# -----------------------------------------------------------------------------
class CosineWarmupScheduler(_LRScheduler):
    """
    Warmup + Cosine Annealing (epoch-aware)
    """

    def __init__(
        self,
        optimizer,
        warmup_epochs,
        max_epochs,
        min_lr=1e-6,
        last_epoch=-1,
    ):
        self.warmup_epochs = warmup_epochs
        self.max_epochs = max_epochs
        self.min_lr = min_lr
        super().__init__(optimizer, last_epoch)

    def get_lr(self):
        step = self.last_epoch + 1

        if step <= self.warmup_epochs:
            return [
                base_lr * step / self.warmup_epochs
                for base_lr in self.base_lrs
            ]

        progress = (step - self.warmup_epochs) / max(
            1, self.max_epochs - self.warmup_epochs
        )

        return [
            self.min_lr
            + (base_lr - self.min_lr)
            * 0.5
            * (1.0 + math.cos(math.pi * progress))
            for base_lr in self.base_lrs
        ]


# -----------------------------------------------------------------------------
# SCHEDULER FACTORY
# -----------------------------------------------------------------------------
def build_scheduler(optimizer, cfg):
    """
    Scheduler principal del proyecto.
    """
    scheduler = CosineWarmupScheduler(
        optimizer=optimizer,
        warmup_epochs=cfg.warmup_epochs,
        max_epochs=cfg.epochs,
        min_lr=cfg.min_lr,
    )
    return scheduler


# -----------------------------------------------------------------------------
# CONFIG DEFAULT (USADO POR CFG GLOBAL)
# -----------------------------------------------------------------------------
class OptimizerConfig:
    lr = 2e-4
    min_lr = 1e-6
    weight_decay = 1e-2
    betas = (0.9, 0.999)
    eps = 1e-8
    warmup_epochs = 5
    epochs = 50


# -----------------------------------------------------------------------------
# SANITY CHECK (DESACTIVADO)
# -----------------------------------------------------------------------------
if False:
    model = torch.nn.Conv2d(3, 1, 3)
    cfg = OptimizerConfig()

    optimizer = build_optimizer(model, cfg)
    scheduler = build_scheduler(optimizer, cfg)

    for epoch in range(cfg.epochs):
        optimizer.step()
        scheduler.step()
        print(epoch, scheduler.get_last_lr()[0])
```

```python
# =================================================================================
# CELL 14 — TRAINING LOOP (DUAL HEAD + METRICS + ROBUSTNESS)
# VESUVIUS CHALLENGE — SURFACE DETECTION
# =================================================================================

import torch
import torch.nn as nn
from torch.cuda.amp import autocast, GradScaler
from tqdm.auto import tqdm
import gc
import os
import numpy as np

# -----------------------------------------------------------------------------
# HELPER: METRIC CALCULATION (DICE SCORE)
# -----------------------------------------------------------------------------
def calculate_dice_score(logits, targets, threshold=0.5):
    """Calcula Dice Score para monitoreo humano durante entrenamiento."""
    probs = torch.sigmoid(logits)
    preds = (probs > threshold).float()
    
    intersection = (preds * targets).sum()
    union = preds.sum() + targets.sum()
    
    # Dice = 2*Int / (Union + epsilon)
    dice = (2.0 * intersection) / (union + 1e-6)
    return dice.item()

# -----------------------------------------------------------------------------
# 1. TRAIN ONE EPOCH
# -----------------------------------------------------------------------------
def train_one_epoch(
    model,
    loader,
    optimizer,
    criterion,
    scaler,
    device,
    ema=None,
    scheduler=None, # Optional: Para OneCycleLR que se actualiza por step
):
    model.train()
    running_loss = 0.0
    running_dice = 0.0
    
    pbar = tqdm(loader, desc="Train", leave=False)

    for batch in pbar:
        # 1. Mover datos a GPU
        # Nota: El Dataset corregido devuelve 'image', 'mask', 'skel'
        images = batch["image"].to(device, non_blocking=True) # [B, Z, C, H, W]
        masks = batch["mask"].to(device, non_blocking=True)   # [B, 1, H, W]
        skels = batch["skel"].to(device, non_blocking=True)   # [B, 1, H, W]

        # 2. Forward Pass (Mixed Precision)
        with autocast(enabled=(device.type == "cuda")):
            # El modelo devuelve tupla: (logits_surf, logits_skel)
            outputs = model(images)
            
            # La Loss Compuesta maneja la tupla internamente
            loss = criterion(outputs, masks, skels)

        # 3. Backward Pass & Optimizacion
        optimizer.zero_grad(set_to_none=True)
        scaler.scale(loss).backward()
        scaler.step(optimizer)
        scaler.update()

        # 4. EMA & Scheduler Update
        if ema is not None:
            ema.update(model)
        
        # Si usas OneCycleLR, el step es por batch
        if scheduler is not None and isinstance(scheduler, torch.optim.lr_scheduler.OneCycleLR):
            scheduler.step()

        # 5. Métricas y Logging
        curr_loss = loss.item()
        running_loss += curr_loss
        
        # Calculamos Dice solo para la cabeza principal (outputs[0] o outputs)
        logits_main = outputs[0] if isinstance(outputs, tuple) else outputs
        curr_dice = calculate_dice_score(logits_main.detach(), masks)
        running_dice += curr_dice

        pbar.set_postfix({"loss": f"{curr_loss:.4f}", "dice": f"{curr_dice:.4f}"})

    # Limpieza de memoria post-epoch
    del images, masks, skels, outputs, loss
    
    avg_loss = running_loss / len(loader)
    avg_dice = running_dice / len(loader)
    
    return avg_loss, avg_dice

# -----------------------------------------------------------------------------
# 2. VALIDATE ONE EPOCH
# -----------------------------------------------------------------------------
@torch.no_grad()
def validate_one_epoch(
    model,
    loader,
    criterion,
    device,
    ema=None,
):
    # Usamos el Context Manager de EMA para validar con los pesos promediados
    # Si ema es None, usamos un contexto vacío (nullcontext sería mejor, pero esto funciona)
    context = EMAContext(ema) if ema is not None else torch.no_grad()
    
    with context:
        model.eval()
        running_loss = 0.0
        running_dice = 0.0
        
        # No usamos tqdm aquí para no ensuciar el log en validación rápida
        for batch in loader:
            images = batch["image"].to(device, non_blocking=True)
            masks = batch["mask"].to(device, non_blocking=True)
            skels = batch["skel"].to(device, non_blocking=True)

            with autocast(enabled=(device.type == "cuda")):
                outputs = model(images)
                loss = criterion(outputs, masks, skels)

            running_loss += loss.item()
            
            logits_main = outputs[0] if isinstance(outputs, tuple) else outputs
            running_dice += calculate_dice_score(logits_main, masks)

    avg_loss = running_loss / len(loader)
    avg_dice = running_dice / len(loader)
    
    return avg_loss, avg_dice

# -----------------------------------------------------------------------------
# 3. UTILS: CHECKPOINT SAVER
# -----------------------------------------------------------------------------
def save_checkpoint(model, optimizer, epoch, score, filename):
    torch.save({
        'epoch': epoch,
        'model_state_dict': model.state_dict(),
        'optimizer_state_dict': optimizer.state_dict(),
        'score': score,
    }, filename)

# -----------------------------------------------------------------------------
# 4. MAIN TRAINING ORCHESTRATOR
# -----------------------------------------------------------------------------
def train_model(
    model,
    train_loader,
    optimizer,
    criterion,
    device,
    epochs,
    scheduler=None,
    val_loader=None,
    ema=None,
    save_dir="checkpoints"
):
    print(f"[*] Iniciando entrenamiento en {device} por {epochs} épocas.")
    model.to(device)
    scaler = GradScaler(enabled=(device.type == "cuda"))
    
    best_dice = 0.0
    history = {"train_loss": [], "val_loss": [], "train_dice": [], "val_dice": []}

    for epoch in range(epochs):
        print(f"\n--- Epoch {epoch+1}/{epochs} ---")
        
        # --- TRAIN ---
        t_loss, t_dice = train_one_epoch(
            model, train_loader, optimizer, criterion, scaler, device, ema, 
            scheduler=(scheduler if isinstance(scheduler, torch.optim.lr_scheduler.OneCycleLR) else None)
        )
        
        # --- VALIDATION ---
        if val_loader:
            v_loss, v_dice = validate_one_epoch(
                model, val_loader, criterion, device, ema
            )
            print(f"    [Train] Loss: {t_loss:.4f} | Dice: {t_dice:.4f}")
            print(f"    [Val  ] Loss: {v_loss:.4f} | Dice: {v_dice:.4f}")
            
            # --- SCHEDULER STEP (Epoch-based) ---
            if scheduler is not None and not isinstance(scheduler, torch.optim.lr_scheduler.OneCycleLR):
                if isinstance(scheduler, torch.optim.lr_scheduler.ReduceLROnPlateau):
                    scheduler.step(v_loss)
                else:
                    scheduler.step()

            # --- CHECKPOINTING ---
            # Guardar último modelo
            save_checkpoint(model, optimizer, epoch, v_dice, os.path.join(save_dir, "last_model.pth"))
            
            # Guardar mejor modelo (basado en Dice)
            if v_dice > best_dice:
                print(f"    [★] New Best Dice: {v_dice:.4f} (was {best_dice:.4f})")
                best_dice = v_dice
                save_checkpoint(model, optimizer, epoch, v_dice, os.path.join(save_dir, "best_model.pth"))
                
                # Si usas EMA, guarda también el modelo EMA como "best_inference"
                if ema:
                    torch.save(ema.ema.state_dict(), os.path.join(save_dir, "best_ema_inference.pth"))

            history["val_loss"].append(v_loss)
            history["val_dice"].append(v_dice)
            
        else:
            print(f"    [Train] Loss: {t_loss:.4f} | Dice: {t_dice:.4f}")

        history["train_loss"].append(t_loss)
        history["train_dice"].append(t_dice)
        
        # Liberar memoria agresivamente entre épocas
        gc.collect()
        torch.cuda.empty_cache()

    return history
```

```python
# =================================================================================
# CELL 15 — VALIDATION LOOP (EMA APPLY)
# VESUVIUS CHALLENGE — SURFACE DETECTION
# =================================================================================

import torch
import numpy as np
from tqdm import tqdm
from contextlib import nullcontext


# -----------------------------------------------------------------------------
# METRICS (SURFACE-AWARE PLACEHOLDERS)
# -----------------------------------------------------------------------------
@torch.no_grad()
def dice_score(pred, target, eps=1e-6):
    pred = (pred > 0.5).float()
    target = target.float()
    inter = (pred * target).sum(dim=(1, 2, 3))
    union = pred.sum(dim=(1, 2, 3)) + target.sum(dim=(1, 2, 3))
    dice = (2 * inter + eps) / (union + eps)
    return dice.mean().item()


@torch.no_grad()
def surface_consistency_penalty(pred):
    gx = torch.abs(pred[..., 1:] - pred[..., :-1]).mean()
    gy = torch.abs(pred[..., 1:, :] - pred[..., :-1, :]).mean()
    return (gx + gy).item()


# -----------------------------------------------------------------------------
# VALIDATION LOOP WITH EMA WEIGHTS
# -----------------------------------------------------------------------------
@torch.no_grad()
def validate_with_ema(
    model,
    ema,
    loader,
    criterion,
    device,
    amp=True,
    return_preds=False,
):
    ema.apply_shadow()
    model.eval()

    total_loss = 0.0
    total_dice = 0.0
    total_surface_penalty = 0.0
    n_batches = 0

    preds_out = [] if return_preds else None

    autocast_ctx = torch.cuda.amp.autocast if amp else nullcontext

    pbar = tqdm(loader, desc="Validation (EMA)", leave=False)

    for batch in pbar:
        images = batch["image"].to(device, non_blocking=True)
        targets = batch["mask"].to(device, non_blocking=True)

        with autocast_ctx():
            logits = model(images)
            loss = criterion(logits, targets)
            probs = torch.sigmoid(logits)

        dice = dice_score(probs, targets)
        surf_pen = surface_consistency_penalty(probs)

        total_loss += loss.item()
        total_dice += dice
        total_surface_penalty += surf_pen
        n_batches += 1

        if return_preds:
            preds_out.append(probs.detach().cpu())

        pbar.set_postfix(
            val_loss=loss.item(),
            dice=dice,
            surf_pen=surf_pen,
        )

    ema.restore()

    results = {
        "val_loss": total_loss / max(1, n_batches),
        "val_dice": total_dice / max(1, n_batches),
        "surface_penalty": total_surface_penalty / max(1, n_batches),
    }

    if return_preds:
        results["predictions"] = torch.cat(preds_out, dim=0)

    return results


# -----------------------------------------------------------------------------
# CHECKPOINT EVALUATION (EMA)
# -----------------------------------------------------------------------------
@torch.no_grad()
def evaluate_checkpoint_ema(
    model,
    ema,
    loader,
    criterion,
    device,
    ckpt_path,
    strict=True,
):
    ckpt = torch.load(ckpt_path, map_location=device)
    model.load_state_dict(ckpt["model"], strict=strict)
    return validate_with_ema(
        model=model,
        ema=ema,
        loader=loader,
        criterion=criterion,
        device=device,
        amp=True,
        return_preds=False,
    )


# -----------------------------------------------------------------------------
# AGGREGATION UTILITIES
# -----------------------------------------------------------------------------
def aggregate_validation_history(history_list):
    keys = history_list[0].keys()
    agg = {}
    for k in keys:
        agg[k] = float(np.mean([h[k] for h in history_list]))
    return agg


# -----------------------------------------------------------------------------
# SANITY CHECK (DISABLED)
# -----------------------------------------------------------------------------
if False:
    device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
    model = torch.nn.Conv2d(3, 1, 3, padding=1).to(device)

    class DummyEMA:
        def apply_shadow(self): pass
        def restore(self): pass

    ema = DummyEMA()

    dummy_loader = []
    validate_with_ema(
        model=model,
        ema=ema,
        loader=dummy_loader,
        criterion=torch.nn.BCEWithLogitsLoss(),
        device=device,
    )
```

```python
# =================================================================================
# CELL 16 — CHECKPOINTING PROFESIONAL
# VESUVIUS CHALLENGE — SURFACE DETECTION
# =================================================================================

import os
import torch
import shutil
from datetime import datetime


# -----------------------------------------------------------------------------
# CHECKPOINT MANAGER
# -----------------------------------------------------------------------------
class CheckpointManager:
    def __init__(
        self,
        save_dir,
        monitor="val_loss",
        mode="min",
        save_top_k=3,
        save_last=True,
        verbose=True,
    ):
        self.save_dir = save_dir
        os.makedirs(self.save_dir, exist_ok=True)

        self.monitor = monitor
        self.mode = mode
        self.save_top_k = save_top_k
        self.save_last = save_last
        self.verbose = verbose

        self.best_k = []
        self.sign = -1 if mode == "max" else 1

    def _is_better(self, score, ref):
        return self.sign * score < self.sign * ref

    def _log(self, msg):
        if self.verbose:
            print(msg)

    def save(
        self,
        epoch,
        model,
        ema,
        optimizer,
        scheduler,
        metrics,
        extra=None,
    ):
        assert self.monitor in metrics, f"Metric '{self.monitor}' not found"

        score = metrics[self.monitor]
        timestamp = datetime.now().strftime("%Y%m%d_%H%M%S")

        ckpt = {
            "epoch": epoch,
            "model": model.state_dict(),
            "ema": ema.state_dict() if ema is not None else None,
            "optimizer": optimizer.state_dict(),
            "scheduler": scheduler.state_dict() if scheduler is not None else None,
            "metrics": metrics,
            "timestamp": timestamp,
        }

        if extra is not None:
            ckpt["extra"] = extra

        filename = f"epoch{epoch:03d}_{self.monitor}={score:.6f}.pth"
        path = os.path.join(self.save_dir, filename)

        # -------------------------
        # SAVE LAST
        # -------------------------
        if self.save_last:
            last_path = os.path.join(self.save_dir, "last.pth")
            torch.save(ckpt, last_path)

        # -------------------------
        # TOP-K LOGIC
        # -------------------------
        if len(self.best_k) < self.save_top_k:
            torch.save(ckpt, path)
            self.best_k.append((score, path))
            self.best_k.sort(key=lambda x: self.sign * x[0])
            self._log(f"[CKPT] Saved: {path}")
        else:
            worst_score, worst_path = self.best_k[-1]
            if self._is_better(score, worst_score):
                torch.save(ckpt, path)
                self._log(f"[CKPT] Replaced: {worst_path} → {path}")
                try:
                    os.remove(worst_path)
                except FileNotFoundError:
                    pass
                self.best_k[-1] = (score, path)
                self.best_k.sort(key=lambda x: self.sign * x[0])

    def save_best_symlink(self):
        if not self.best_k:
            return
        best_path = self.best_k[0][1]
        link_path = os.path.join(self.save_dir, "best.pth")
        try:
            if os.path.islink(link_path) or os.path.exists(link_path):
                os.remove(link_path)
            shutil.copy(best_path, link_path)
            self._log(f"[CKPT] Best checkpoint updated → {best_path}")
        except Exception:
            pass


# -----------------------------------------------------------------------------
# LOAD CHECKPOINT (SAFE)
# -----------------------------------------------------------------------------
def load_checkpoint(
    path,
    model,
    optimizer=None,
    scheduler=None,
    ema=None,
    device="cpu",
    strict=True,
):
    ckpt = torch.load(path, map_location=device)

    model.load_state_dict(ckpt["model"], strict=strict)

    if optimizer is not None and "optimizer" in ckpt:
        optimizer.load_state_dict(ckpt["optimizer"])

    if scheduler is not None and "scheduler" in ckpt and ckpt["scheduler"] is not None:
        scheduler.load_state_dict(ckpt["scheduler"])

    if ema is not None and "ema" in ckpt and ckpt["ema"] is not None:
        ema.load_state_dict(ckpt["ema"])

    epoch = ckpt.get("epoch", None)
    metrics = ckpt.get("metrics", None)

    return epoch, metrics


# -----------------------------------------------------------------------------
# SANITY CHECK (DISABLED)
# -----------------------------------------------------------------------------
if False:
    manager = CheckpointManager(
        save_dir="./checkpoints",
        monitor="val_loss",
        mode="min",
        save_top_k=2,
    )
```

```python
# =================================================================================
# CELL 17 — INFERENCIA FULL VOLUME (2.5D SLIDING WINDOW + TILING)
# VESUVIUS CHALLENGE — SURFACE DETECTION
# =================================================================================

import os
import torch
import numpy as np
import tifffile
import zarr
import gc
from torch.utils.data import Dataset, DataLoader
from tqdm.auto import tqdm
from contextlib import contextmanager, nullcontext

# -----------------------------------------------------------------------------
# UTILS
# -----------------------------------------------------------------------------
@contextmanager
def inference_mode():
    with torch.no_grad():
        yield

def sigmoid(x):
    return 1 / (1 + np.exp(-x))

# -----------------------------------------------------------------------------
# DATASET DE INFERENCIA (SLIDING WINDOW 2.5D)
# -----------------------------------------------------------------------------
class InferenceSlidingWindowDataset(Dataset):
    """
    Recorre el volumen completo extrayendo parches 3D [Z, C, H, W]
    para alimentar al modelo VesuviusSurfaceNet.
    """
    def __init__(
        self,
        feature_volume,          # [Z, C, H, W] (Zarr o Memmap)
        tile_size=256,
        stride=128,
        z_window=16,
    ):
        self.volume = feature_volume
        self.tile_size = tile_size
        self.stride = stride
        self.z_window = z_window
        self.half_z = z_window // 2
        
        # Dimensiones del volumen
        self.Z = feature_volume.shape[0]
        self.H = feature_volume.shape[2]
        self.W = feature_volume.shape[3]
        
        # Generar coordenadas de tiles (y, x)
        self.coords = []
        for y in range(0, self.H, stride):
            for x in range(0, self.W, stride):
                # Ajuste para el último tile si se sale del borde
                y_eff = min(y, self.H - tile_size)
                x_eff = min(x, self.W - tile_size)
                self.coords.append((y_eff, x_eff))
                
        # Eliminamos duplicados que genera el ajuste de bordes
        self.coords = sorted(list(set(self.coords)))

    def __len__(self):
        return len(self.coords)

    def __getitem__(self, idx):
        y, x = self.coords[idx]
        
        # ESTRATEGIA DE INFERENCIA 2.5D:
        # Predecimos la superficie asumiendo que está cerca del centro Z del volumen
        # o iteramos sobre Z si no sabemos dónde está.
        # Para Vesuvius Surface Detection, normalmente escaneamos todo Z o usamos
        # un "prior" de altura. Aquí asumimos escaneo central robusto o pre-calculado.
        
        # Opción A: Usamos el centro geométrico del volumen (Simple baseline)
        center_z = self.Z // 2
        
        z_start = max(0, center_z - self.half_z)
        z_end = min(self.Z, center_z + self.half_z)
        
        # Padding si estamos en los bordes Z
        pad_pre = max(0, self.half_z - (center_z - z_start))
        pad_post = max(0, self.half_z - (z_end - center_z))
        
        # Extraer parche [Z_actual, C, H, W]
        patch = self.volume[z_start:z_end, :, y : y + self.tile_size, x : x + self.tile_size]
        
        # Convertir a float32
        patch = patch.astype(np.float32)
        
        # Padding en Z si es necesario para mantener tamaño fijo
        if pad_pre > 0 or pad_post > 0:
            patch = np.pad(patch, ((pad_pre, pad_post), (0,0), (0,0), (0,0)), mode='reflect')
            
        # Tensorizar
        return torch.from_numpy(patch), y, x

# -----------------------------------------------------------------------------
# STITCHER (RECONSTRUCTOR DE IMAGEN)
# -----------------------------------------------------------------------------
class ImageStitcher:
    """Acumula predicciones superpuestas y promedia los resultados."""
    def __init__(self, H, W):
        self.H = H
        self.W = W
        self.prob_map = np.zeros((H, W), dtype=np.float32)
        self.count_map = np.zeros((H, W), dtype=np.float32)

    def add(self, prob_patch, y, x):
        """prob_patch: [tile_size, tile_size]"""
        h_p, w_p = prob_patch.shape
        
        # Acumular suma y conteo
        self.prob_map[y : y + h_p, x : x + w_p] += prob_patch
        self.count_map[y : y + h_p, x : x + w_p] += 1.0

    def finalize(self):
        # Evitar división por cero
        return self.prob_map / np.maximum(self.count_map, 1.0)

# -----------------------------------------------------------------------------
# INFERENCE ENGINE (CORE)
# -----------------------------------------------------------------------------
def infer_full_volume(
    model,
    feature_volume, # Zarr/Array [Z, C, H, W] pre-calculado
    tile_size=256,
    stride=128,
    z_window=16,
    batch_size=4,
    device="cuda",
    ema=None,
):
    # 1. Preparar Modelo
    model.to(device)
    model.eval()
    
    # Contexto EMA seguro
    context = EMAContext(ema) if ema is not None else nullcontext()

    # 2. Dataset y Loader
    dataset = InferenceSlidingWindowDataset(
        feature_volume=feature_volume,
        tile_size=tile_size,
        stride=stride,
        z_window=z_window
    )
    
    loader = DataLoader(
        dataset,
        batch_size=batch_size,
        shuffle=False,
        num_workers=2,
        pin_memory=True
    )
    
    # 3. Stitcher
    stitcher = ImageStitcher(dataset.H, dataset.W)
    
    print(f"[*] Iniciando Inferencia: {len(dataset)} tiles...")
    
    with context:
        with torch.no_grad():
            for batch_imgs, batch_ys, batch_xs in tqdm(loader, desc="Inference"):
                batch_imgs = batch_imgs.to(device, non_blocking=True) # [B, Z, C, H, W]
                
                # Forward (Mixed Precision opcional pero recomendado)
                with torch.cuda.amp.autocast(enabled=True):
                    outputs = model(batch_imgs)
                    
                    # Unpack dual head: (logits, skel) o solo logits
                    if isinstance(outputs, tuple):
                        logits = outputs[0]
                    else:
                        logits = outputs
                
                # Sigmoid y a CPU
                probs = torch.sigmoid(logits).detach().cpu().numpy() # [B, 1, H, W]
                probs = probs.squeeze(1) # [B, H, W]
                
                # Stitching
                for i in range(len(probs)):
                    y = batch_ys[i].item()
                    x = batch_xs[i].item()
                    stitcher.add(probs[i], y, x)
                    
    return stitcher.finalize()

# -----------------------------------------------------------------------------
# PIPELINE WRAPPER (END-TO-END)
# -----------------------------------------------------------------------------
def run_inference_pipeline(
    model,
    volume_id,
    volume_path_dict, # Diccionario de rutas del vol (tif/zarr)
    output_dir,
    device="cuda",
    ema=None,
):
    print(f"[*] Procesando Volumen ID: {volume_id}")
    
    # 1. Cargar Volumen Raw (Lazy)
    raw_vol = safe_load_volume(volume_path_dict, volume_id)
    
    # 2. Generar Features (On-the-fly o Cached)
    # Nota: Para inferencia, si no tenemos RAM para todo el feature_vol,
    # deberíamos generar features por tile. Aquí asumimos que cabe o usamos Zarr en disco.
    # Para Kaggle, lo ideal es crear un Zarr temporal en disco si no cabe en RAM.
    
    print("[*] Generando Features (esto puede tardar)...")
    # Simplificación: Usamos una función que devuelva un Zarr array en disco
    # Si no tienes esa función, usa build_feature_volume pero cuidado con la RAM.
    # Aquí asumimos que build_feature_volume puede manejar chunks o usamos raw para test.
    
    # HACK para Demo: Usamos raw_vol expandido si no podemos computar features rápido
    # En producción real: feature_vol, _ = build_feature_volume(raw_vol, ...)
    # Adaptador temporal:
    feature_vol = raw_vol # Placeholder si no se corre el feature eng completo
    if feature_vol.ndim == 3:
        feature_vol = feature_vol[:, np.newaxis, :, :] # [Z, 1, H, W]
    
    # 3. Ejecutar Inferencia
    pred_map = infer_full_volume(
        model=model,
        feature_volume=feature_vol,
        tile_size=PATCH_SIZE,
        stride=PATCH_STRIDE,
        z_window=NUM_SLICES_CONTEXT,
        batch_size=BATCH_SIZE * 2, # Batch size doble para inferencia
        device=device,
        ema=ema
    )
    
    # 4. Guardar Resultado
    os.makedirs(output_dir, exist_ok=True)
    out_path = os.path.join(output_dir, f"{volume_id}_prediction.tif")
    
    # Guardar como uint16 para precisión
    pred_uint16 = (pred_map * 65535).astype(np.uint16)
    tifffile.imwrite(out_path, pred_uint16)
    
    print(f"[✅] Predicción guardada en: {out_path}")
    return pred_map, out_path
```

```python
# =================================================================================
# CELL 18 — POSTPROCESADO FINAL
# VESUVIUS CHALLENGE — SURFACE DETECTION
# =================================================================================

import os
import numpy as np
import tifffile as tiff
import cv2
from scipy import ndimage as ndi
from skimage.morphology import (
    disk,
    binary_opening,
    binary_closing,
    remove_small_objects,
    remove_small_holes,
)
from skimage.filters import threshold_otsu
from skimage.exposure import rescale_intensity


# -----------------------------------------------------------------------------
# NORMALIZATION
# -----------------------------------------------------------------------------
def normalize_01(x, eps=1e-8):
    x = x.astype(np.float32)
    return (x - x.min()) / (x.max() - x.min() + eps)


def normalize_percentile(x, p_low=1.0, p_high=99.0):
    lo, hi = np.percentile(x, [p_low, p_high])
    return np.clip((x - lo) / (hi - lo + 1e-6), 0, 1)


# -----------------------------------------------------------------------------
# SMOOTHING
# -----------------------------------------------------------------------------
def smooth_map(x, gaussian_sigma=1.0, median_ksize=0):
    if gaussian_sigma > 0:
        x = ndi.gaussian_filter(x, sigma=gaussian_sigma)
    if median_ksize > 1:
        x = cv2.medianBlur((x * 255).astype(np.uint8), median_ksize) / 255.0
    return x


# -----------------------------------------------------------------------------
# THRESHOLDING
# -----------------------------------------------------------------------------
def adaptive_threshold(x, method="otsu", fixed_threshold=0.5):
    if method == "otsu":
        t = threshold_otsu(x)
        return x > t
    elif method == "fixed":
        return x > fixed_threshold
    else:
        raise ValueError("Unknown threshold method")


# -----------------------------------------------------------------------------
# MORPHOLOGICAL CLEANING
# -----------------------------------------------------------------------------
def clean_mask(
    mask,
    min_object_size=64,
    min_hole_size=64,
    open_radius=2,
    close_radius=2,
):
    if open_radius > 0:
        mask = binary_opening(mask, disk(open_radius))
    if close_radius > 0:
        mask = binary_closing(mask, disk(close_radius))

    mask = remove_small_objects(mask, min_size=min_object_size)
    mask = remove_small_holes(mask, area_threshold=min_hole_size)

    return mask


# -----------------------------------------------------------------------------
# CONNECTED COMPONENT FILTERING
# -----------------------------------------------------------------------------
def filter_components(mask, min_area=128, max_area=None):
    labeled, num = ndi.label(mask)
    if num == 0:
        return mask

    sizes = ndi.sum(mask, labeled, range(1, num + 1))
    filtered = np.zeros_like(mask, dtype=bool)

    for i, size in enumerate(sizes, start=1):
        if size < min_area:
            continue
        if max_area is not None and size > max_area:
            continue
        filtered[labeled == i] = True

    return filtered


# -----------------------------------------------------------------------------
# EDGE-AWARE REFINEMENT
# -----------------------------------------------------------------------------
def edge_refine(prob, mask, edge_sigma=1.0):
    grad_x = ndi.sobel(prob, axis=1)
    grad_y = ndi.sobel(prob, axis=0)
    edges = np.hypot(grad_x, grad_y)
    edges = ndi.gaussian_filter(edges, sigma=edge_sigma)
    edges = normalize_01(edges)

    refined = mask & (edges < np.percentile(edges, 85))
    return refined


# -----------------------------------------------------------------------------
# FINAL POSTPROCESS PIPELINE
# -----------------------------------------------------------------------------
def postprocess_prediction(
    prob_map,
    normalize_mode="percentile",
    gaussian_sigma=1.0,
    median_ksize=0,
    threshold_method="otsu",
    fixed_threshold=0.5,
    min_object_size=96,
    min_hole_size=96,
    open_radius=2,
    close_radius=2,
    component_min_area=128,
    component_max_area=None,
    edge_refinement=True,
):
    x = prob_map.copy()

    if normalize_mode == "percentile":
        x = normalize_percentile(x)
    elif normalize_mode == "minmax":
        x = normalize_01(x)

    x = smooth_map(x, gaussian_sigma, median_ksize)

    mask = adaptive_threshold(
        x, method=threshold_method, fixed_threshold=fixed_threshold
    )

    mask = clean_mask(
        mask,
        min_object_size=min_object_size,
        min_hole_size=min_hole_size,
        open_radius=open_radius,
        close_radius=close_radius,
    )

    mask = filter_components(
        mask,
        min_area=component_min_area,
        max_area=component_max_area,
    )

    if edge_refinement:
        mask = edge_refine(x, mask)

    return x, mask


# -----------------------------------------------------------------------------
# IO HELPERS
# -----------------------------------------------------------------------------
def save_outputs(
    prob_map,
    mask,
    out_dir,
    base_name,
):
    os.makedirs(out_dir, exist_ok=True)

    prob_uint16 = np.clip(prob_map * 65535.0, 0, 65535).astype(np.uint16)
    mask_uint8 = (mask.astype(np.uint8) * 255)

    tiff.imwrite(os.path.join(out_dir, f"{base_name}_prob.tif"), prob_uint16)
    tiff.imwrite(os.path.join(out_dir, f"{base_name}_mask.tif"), mask_uint8)


# -----------------------------------------------------------------------------
# FULL POSTPROCESS ENTRYPOINT
# -----------------------------------------------------------------------------
def run_postprocess(
    prediction_path,
    output_dir,
    base_name="vesuvius",
    **kwargs,
):
    prob_map = tiff.imread(prediction_path).astype(np.float32)
    prob_map = normalize_01(prob_map)

    prob_pp, mask_pp = postprocess_prediction(prob_map, **kwargs)

    save_outputs(prob_pp, mask_pp, output_dir, base_name)

    return prob_pp, mask_pp


# -----------------------------------------------------------------------------
# SANITY CHECK (DISABLED)
# -----------------------------------------------------------------------------
if False:
    prob, mask = run_postprocess(
        prediction_path="pred_scroll_1.tif",
        output_dir="./postprocess",
        base_name="scroll_1",
        gaussian_sigma=1.2,
        threshold_method="otsu",
        min_object_size=128,
        component_min_area=256,
    )
```

```python
# =================================================================================
# CELL 19 — GENERACIÓN DE SUBMISSION
# VESUVIUS CHALLENGE — SURFACE DETECTION
# =================================================================================

import os
import numpy as np
import tifffile as tiff
import pandas as pd
from pathlib import Path
from tqdm import tqdm
from skimage.transform import resize


# -----------------------------------------------------------------------------
# CONFIGURACIÓN GLOBAL
# -----------------------------------------------------------------------------
SUBMISSION_VERSION = "v1_final"
DEFAULT_THRESHOLD = 0.5
TARGET_DTYPE = np.uint8


# -----------------------------------------------------------------------------
# UTILS
# -----------------------------------------------------------------------------
def ensure_dir(path):
    path = Path(path)
    path.mkdir(parents=True, exist_ok=True)
    return path


def normalize_01(x, eps=1e-8):
    x = x.astype(np.float32)
    return (x - x.min()) / (x.max() - x.min() + eps)


def resize_if_needed(mask, target_shape, order=0):
    if mask.shape == target_shape:
        return mask
    return resize(
        mask,
        target_shape,
        order=order,
        preserve_range=True,
        anti_aliasing=False,
    )


# -----------------------------------------------------------------------------
# LOAD FINAL MASK
# -----------------------------------------------------------------------------
def load_final_mask(mask_path):
    mask = tiff.imread(mask_path)
    if mask.dtype != np.uint8:
        mask = (mask > 0).astype(np.uint8)
    return mask


# -----------------------------------------------------------------------------
# PIXEL EXTRACTION
# -----------------------------------------------------------------------------
def mask_to_pixels(mask):
    h, w = mask.shape
    ys, xs = np.nonzero(mask)
    pixels = ys * w + xs
    return pixels.astype(np.int64)


# -----------------------------------------------------------------------------
# BUILD SUBMISSION DATAFRAME
# -----------------------------------------------------------------------------
def build_submission_df(scroll_id, mask):
    pixels = mask_to_pixels(mask)
    df = pd.DataFrame(
        {
            "id": np.repeat(scroll_id, len(pixels)),
            "pixel": pixels,
        }
    )
    return df


# -----------------------------------------------------------------------------
# MULTI-SCROLL PIPELINE
# -----------------------------------------------------------------------------
def generate_submission(
    masks_root,
    output_dir,
    scroll_ids,
    target_shapes=None,
    submission_name="submission.csv",
):
    masks_root = Path(masks_root)
    output_dir = ensure_dir(output_dir)

    all_dfs = []

    for i, scroll_id in enumerate(tqdm(scroll_ids, desc="Generating submission")):
        mask_path = masks_root / f"{scroll_id}_mask.tif"
        assert mask_path.exists(), f"Missing mask: {mask_path}"

        mask = load_final_mask(mask_path)

        if target_shapes is not None:
            mask = resize_if_needed(
                mask,
                target_shapes[i],
                order=0,
            )

        df = build_submission_df(scroll_id, mask)
        all_dfs.append(df)

    submission_df = pd.concat(all_dfs, ignore_index=True)

    submission_path = output_dir / submission_name
    submission_df.to_csv(submission_path, index=False)

    return submission_path, submission_df


# -----------------------------------------------------------------------------
# SINGLE-SCROLL SHORTCUT
# -----------------------------------------------------------------------------
def generate_single_scroll_submission(
    mask_path,
    scroll_id,
    output_dir,
    submission_name="submission.csv",
):
    output_dir = ensure_dir(output_dir)

    mask = load_final_mask(mask_path)
    df = build_submission_df(scroll_id, mask)

    submission_path = output_dir / submission_name
    df.to_csv(submission_path, index=False)

    return submission_path, df


# -----------------------------------------------------------------------------
# OPTIONAL VALIDATION
# -----------------------------------------------------------------------------
def validate_submission(df):
    assert "id" in df.columns
    assert "pixel" in df.columns
    assert df["pixel"].dtype in (np.int64, np.int32)
    assert len(df) > 0
    assert df["pixel"].min() >= 0
    return True


# -----------------------------------------------------------------------------
# ENTRYPOINT (DISABLED)
# -----------------------------------------------------------------------------
if False:
    submission_path, submission_df = generate_submission(
        masks_root="./postprocess",
        output_dir="./submission",
        scroll_ids=["scroll_1"],
        target_shapes=[(1024, 1024)],
        submission_name=f"vesuvius_{SUBMISSION_VERSION}.csv",
    )

    validate_submission(submission_df)
```

```python
# =================================================================================
# CELL 20 — VISUALIZACIÓN & DEBUG
# VESUVIUS CHALLENGE — SURFACE DETECTION
# =================================================================================

import numpy as np
import matplotlib.pyplot as plt
import cv2
import tifffile as tiff
from pathlib import Path
from skimage.morphology import skeletonize, dilation, disk
from skimage.exposure import rescale_intensity


# -----------------------------------------------------------------------------
# UTILIDADES BASE
# -----------------------------------------------------------------------------
def load_tif_any(path):
    arr = tiff.imread(path)
    if arr.ndim == 3:
        arr = arr[arr.shape[0] // 2]
    return arr


def norm_uint8(x):
    x = x.astype(np.float32)
    x = rescale_intensity(x, in_range="image", out_range=(0, 255))
    return x.astype(np.uint8)


def overlay_mask(img, mask, color=(255, 0, 0), alpha=0.4):
    overlay = img.copy()
    colored = np.zeros_like(img)
    colored[..., 0] = color[0]
    colored[..., 1] = color[1]
    colored[..., 2] = color[2]
    overlay[mask > 0] = (
        overlay[mask > 0] * (1 - alpha) + colored[mask > 0] * alpha
    )
    return overlay.astype(np.uint8)


# -----------------------------------------------------------------------------
# VISUALIZACIÓN DE SLICE + PREDICCIÓN
# -----------------------------------------------------------------------------
def visualize_slice_prediction(
    volume_path,
    pred_mask_path,
    slice_idx=None,
    figsize=(12, 12),
):
    vol = tiff.imread(volume_path)
    pred = tiff.imread(pred_mask_path)

    if slice_idx is None:
        slice_idx = vol.shape[0] // 2

    img = norm_uint8(vol[slice_idx])
    mask = pred.astype(bool)

    img_rgb = cv2.cvtColor(img, cv2.COLOR_GRAY2RGB)
    overlay = overlay_mask(img_rgb, mask, color=(255, 0, 0), alpha=0.45)

    plt.figure(figsize=figsize)
    plt.imshow(overlay)
    plt.title(f"Slice {slice_idx} — Prediction Overlay")
    plt.axis("off")
    plt.show()


# -----------------------------------------------------------------------------
# VISUALIZACIÓN SKELETON + SUPERFICIE
# -----------------------------------------------------------------------------
def visualize_skeleton_alignment(
    image,
    skeleton,
    surface_mask,
    figsize=(14, 10),
):
    img = norm_uint8(image)
    img_rgb = cv2.cvtColor(img, cv2.COLOR_GRAY2RGB)

    skel_layer = np.zeros_like(img_rgb)
    skel_layer[skeleton > 0] = (0, 255, 0)
    skel_layer = cv2.dilate(skel_layer, np.ones((3, 3), np.uint8))

    surf_layer = np.zeros_like(img_rgb)
    surf_layer[surface_mask > 0] = (255, 0, 0)

    combined = cv2.addWeighted(img_rgb, 0.6, skel_layer, 0.7, 0)
    combined = cv2.addWeighted(combined, 1.0, surf_layer, 0.5, 0)

    plt.figure(figsize=figsize)
    plt.imshow(combined)
    plt.title("Skeleton (Green) vs Surface Prediction (Red)")
    plt.axis("off")
    plt.show()


# -----------------------------------------------------------------------------
# DEBUG MAPAS INTERMEDIOS
# -----------------------------------------------------------------------------
def debug_maps(
    base_image,
    gravity_map=None,
    probability_map=None,
    skeleton=None,
    figsize=(18, 6),
):
    n = 1 + sum(x is not None for x in [gravity_map, probability_map, skeleton])
    plt.figure(figsize=figsize)

    idx = 1
    plt.subplot(1, n, idx)
    plt.imshow(norm_uint8(base_image), cmap="gray")
    plt.title("Base Image")
    plt.axis("off")
    idx += 1

    if gravity_map is not None:
        plt.subplot(1, n, idx)
        plt.imshow(gravity_map, cmap="inferno")
        plt.title("Gravity Map")
        plt.axis("off")
        idx += 1

    if probability_map is not None:
        plt.subplot(1, n, idx)
        plt.imshow(probability_map, cmap="magma")
        plt.title("Probability Map")
        plt.axis("off")
        idx += 1

    if skeleton is not None:
        plt.subplot(1, n, idx)
        plt.imshow(skeleton, cmap="gray")
        plt.title("Skeleton")
        plt.axis("off")

    plt.tight_layout()
    plt.show()


# -----------------------------------------------------------------------------
# QUICK SANITY CHECK
# -----------------------------------------------------------------------------
def sanity_check_prediction(mask):
    total = mask.size
    positive = mask.sum()
    ratio = positive / total
    print(f"[SANITY] Positive pixels: {positive} / {total} ({ratio:.6f})")
    return ratio


# -----------------------------------------------------------------------------
# ENTRYPOINT DE DEBUG (DESACTIVADO)
# -----------------------------------------------------------------------------
if False:
    vol_path = "/kaggle/input/vesuvius-challenge-surface-detection/train_images/1004283650.tif"
    pred_path = "./postprocess/1004283650_mask.tif"

    visualize_slice_prediction(
        volume_path=vol_path,
        pred_mask_path=pred_path,
        slice_idx=500,
    )

    img = load_tif_any(vol_path)
    skel = skeletonize(img > img.mean())
    mask = tiff.imread(pred_path)

    visualize_skeleton_alignment(
        image=img,
        skeleton=skel,
        surface_mask=mask,
    )

    sanity_check_prediction(mask)
```

```python
# =================================================================================
# HOTFIX: REPARACIÓN DE CONFIGURACIÓN DEL OPTIMIZADOR
# =================================================================================
from dataclasses import dataclass
import torch.optim as optim

# 1. Redefinimos la configuración correctamente como Dataclass
@dataclass
class OptimizerConfig:
    lr: float = 2e-4
    weight_decay: float = 1e-2
    eta_min: float = 1e-6
    epochs: int = 25

# 2. Aseguramos que los builders existan también
def build_optimizer(model, cfg: OptimizerConfig):
    return optim.AdamW(
        model.parameters(), 
        lr=cfg.lr, 
        weight_decay=cfg.weight_decay
    )

def build_scheduler(optimizer, cfg: OptimizerConfig):
    return optim.lr_scheduler.CosineAnnealingLR(
        optimizer,
        T_max=cfg.epochs,
        eta_min=cfg.eta_min,
    )

print("✅ OptimizerConfig reparado. Ahora puedes ejecutar la Celda 24.")
```
