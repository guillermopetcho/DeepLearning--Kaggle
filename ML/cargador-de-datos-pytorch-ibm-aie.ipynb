<h2 id="data_class">Create a Dataset Class</h2>


In this section, we will use the previous code to build a dataset class. 



Complete the code to build a Dataset class <code>dataset</code>. As before, make sure the even samples are positive, and the odd samples are negative.  If the parameter <code>train</code> is set to <code>True</code>, use the first 30 000  samples as training data; otherwise, the remaining samples will be used as validation data.  


```python
import os
from PIL import Image
import torch
from torch.utils.data import Dataset as _Dataset

class Dataset(_Dataset):

    def __init__(self, transform=None, train=True):
        #  ──> Ajusta aquí la ruta base a la carpeta que realmente existe en tu entorno:
        directory = "./data"
        positive  = "Positive"
        negative  = "Negative"

        # Construimos las rutas absolutas a Positive/ y Negative/
        positive_file_path = os.path.join(directory, positive)
        negative_file_path = os.path.join(directory, negative)

        # Listamos y ordenamos todos los .jpg de cada carpeta
        positive_files = [
            os.path.join(positive_file_path, fname)
            for fname in os.listdir(positive_file_path)
            if fname.endswith(".jpg")
        ]
        positive_files.sort()

        negative_files = [
            os.path.join(negative_file_path, fname)
            for fname in os.listdir(negative_file_path)
            if fname.endswith(".jpg")
        ]
        negative_files.sort()

        # Guardamos el transformador (si se pasa)
        self.transform = transform

        # Número total de imágenes
        number_of_samples = len(positive_files) + len(negative_files)

        # Preparamos la lista intercalada: pares=positivas, impares=negativas
        self.all_files = [None] * number_of_samples
        for i, pos_path in enumerate(positive_files):
            self.all_files[2 * i] = pos_path
        for i, neg_path in enumerate(negative_files):
            self.all_files[2 * i + 1] = neg_path

        # Creamos el tensor de etiquetas (1=Positive en índices pares, 0=Negative en impares)
        Y = torch.zeros(number_of_samples, dtype=torch.long)
        Y[::2] = 1
        Y[1::2] = 0

        # Split Train/Validation
        if train:
            self.all_files = self.all_files[:30000]
            self.Y         = Y[:30000]
        else:
            self.all_files = self.all_files[30000:]
            self.Y         = Y[30000:]

        self.len = len(self.all_files)

    def __len__(self):
        return self.len

    def __getitem__(self, idx):
        img = Image.open(self.all_files[idx]).convert("RGB")
        label = self.Y[idx]
        if self.transform is not None:
            img = self.transform(img)
        return img, label
```

<h2 id="Question_4">Question 4</h2>
<b>Create a Dataset object <code>dataset</code> for the training data, use the following lines of code to print out sample the 10th and  sample 100 (remember zero indexing)  </b>


```python
# Celda 45
# Question 4: Mostrar muestras de entrenamiento
# Por ejemplo, usando un transform simple:
from torchvision import transforms

transform_simple = transforms.Compose([
    transforms.Resize((28, 28)),
    transforms.ToTensor()
])

dataset = Dataset(transform=transform_simple, train=True)
samples = [10, 100]  # Recuerda que es 0-indexed

for sample in samples:
    img, lbl = dataset[sample]
    plt.imshow(img.permute(1, 2, 0), cmap='gray')  # si ToTensor(), permutamos canales
    plt.xlabel(f"y = {lbl.item()}")
    plt.title(f"training data, sample {sample}")
    plt.axis('off')
    plt.show()
```
